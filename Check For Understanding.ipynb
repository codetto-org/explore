# Check For Understanding

**Check for Understanding cells** let you place a quick question directly inside a notebook. Students answer it and find out straight away whether they were right. This can be useful for confirming that an idea has landed before students move on to the next part of an activity.

## Why use a Check for Understanding cell?

- **Keeps everything in one place.** No switching between a notebook and a separate quiz tool.
- **Instant feedback.** Students see whether they were right as soon as they submit, with nothing for you to mark.
- **Low stakes.** A wrong answer shows the correct one, and students can try again as often as they like.

## The source format

A Check for Understanding cell's source is a small block of settings, written as JSON:

```
{
  "question_type": "true_false",
  "question": "Python is a programming language.",
  "answer": "True",
  "submitted_answer": ""
}
```

The available settings:

- `question_type`: `multiple_choice`, `true_false`, or `freeform`. See **Question types** below.
- `question`: the question shown to the student.
- `answer`: the correct answer.
- `options`: the list of choices. Only needed for `multiple_choice` questions.
- `submitted_answer`: where the student's answer is stored. Leave it as `""` when writing a question.
- `animation`: `true` or `false` (default `true`). Set it to `false` to turn off the confetti shown for a correct answer.

The first three are required. JSON is strict about punctuation: text values need double quotes, `true` and `false` don't, and every setting except the last ends with a comma.

## Question types

**Multiple choice** questions list their choices under `options`. Each one has a short `key` and the `text` students see, and `answer` is the key of the correct choice:

```
{
  "question_type": "multiple_choice",
  "question": "Which of these is a loop?",
  "options": [
    { "key": "a", "text": "if" },
    { "key": "b", "text": "for" },
    { "key": "c", "text": "print" }
  ],
  "answer": "b",
  "submitted_answer": ""
}
```

You can have as many or as few choices as you like. The example at the top of this notebook is a multiple choice question.

**True / False** questions show a statement for students to judge, so write the `question` as a statement. The `answer` is either `"True"` or `"False"`. Try the example below.

**Free response** questions (`freeform`) give students a box to type into. The student's answer has to match `answer` exactly, although capital letters and extra spaces at either end are ignored. With an `answer` of `"json"`, both `JSON` and ` Json ` are correct, but `a json file` is not.

This makes free response best for questions with one short, unambiguous answer, such as a single word or a number. Try the example below.

## Feedback and trying again

A correct answer shows **Correct!** and a burst of confetti. A wrong answer shows **Not quite.** along with the correct answer.

Because the correct answer is revealed, these questions work best as a quick check during a lesson, and not as a test. After answering, students can click **Try again** to clear their answer and have another go.

If the confetti would be a distraction, add `"animation": false` to the question. The example below has it turned off.

## Editing tools

To add a question, go into edit mode (the pencil icon on the toolbar), click **Add cell** where you'd like the question to go, choose **Check for Understanding**, and pick a question type. The new cell starts with a sample question, ready for you to replace.

To change a question, click the edit icon in its top-right corner. This opens the settings as text. Replace the `question` and `answer` (and the `options`, for multiple choice), then click **Close** to save.

If you see a red error message instead of the question, the settings couldn't be read. Check that `question_type`, `question`, and `answer` are all present, and look for a missing quote or comma.

## Tracking progress

Each question is labelled with its place in the notebook, such as **Question 2 of 4**. The numbering is automatic, so you can add or reorder questions freely.

Questions are also listed in the **Notebook Map**, the panel on the right. Each one shows its number until it's answered, then whether the answer was **Correct**. The **Next Question** button at the bottom of the map jumps to the next question that hasn't been answered yet.

Answers are saved with the notebook. If you've tried the questions yourself and want to hand the notebook out, export it and choose **Export a clean copy**, which clears every answer.
