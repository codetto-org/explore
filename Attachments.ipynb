# Using Attachments

**Attachments** are external files that are attached to a notebook, similar to an attachment to an email message. An attachment can be any time of file: an image, a sound clip, a CSV file, or anything your code might need. Once attached, a file is available to Python at a fixed location: `/attachments/<filename>`.

Unlike environment variables, which stay private to one device, attachments travel *with* the notebook. Export it, share it, or hand it to a student, and the attachments go along for the ride.

## Why attach files?

- **Give students a starting dataset.** Attach a CSV of real data and every student who opens the notebook can analyze it immediately with no separate upload step.
- **Bundle project assets.** A starter image, a sound effect, a background texture for a game. Attach it once and it becomes part of the notebook.
- **Save your own data as an attachment.** Student code can create new files, saved as attachments linked to the notebook. Examples could include a generated dataset, a saved high-score list, or any other output from the code.

This notebook already has a file attached. Let's look at it!

## Look at what's already attached

Click the paperclip icon in the top right to open the **Attachments** panel. You'll see `sample_data.csv` already listed, along with its size.

Now let's read that file from Python.

```python
with open('/attachments/sample_data.csv') as f:
  print(f.read())
```

## Adding your own file

To attach a file of your own:

1. Open the **Attachments** panel (paperclip icon).
2. Click **Add File** and choose a file from your computer.
3. The attachment is now available at `/attachments/<name>` the next time a cell runs. Change the name in the previous cell to test it out.

Note: To keep notebooks portable, attachments have limits: **5 MB per file**, **20 MB total** across every file in a notebook.

## Code can create files too

Python code doesn't just *read* from `/attachments` - it can also *write* to them as well. Writing a new file, or changing an existing one, saves it back into the notebook automatically.

Run the cell below, then open the Attachments panel again.

```python
with open('/attachments/summary.txt', 'w') as f:
  f.write('Generated by code!\n')
  f.write('This file is saved with the notebook, just like an uploaded one.\n')

print('Wrote /attachments/summary.txt')
```

`summary.txt` now shows up in the Attachments panel. No upload is needed. It will still be there the next time this notebook is opened, and for others if you share this notebook.

Attachments created using code are updated when the cell finishes executing. A file too large to save permanently (over the 5 MB / 20 MB limits above) stays available for the rest of the current session, but won't be retained once you leave.
