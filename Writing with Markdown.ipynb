# Writing with Markdown

**Markdown cells** are the text cells within a notebook. These are used for instructions, explanations, and headings. This guide covers the basics of writing and formatting markdown, adding images, and available editor options.

**Tip**: To edit any markdown cell in notebook, go into edit mode (the pencil icon on the toolbar), and double-click the markdown cell or click on the edit icon. Try editing this cell now, then click Close to leave it unchanged.

## Why formatting matters

- **Headings organize a notebook.** A level-1 heading (`#`) starts a new major section. Codetto turns every one of them into a collapsible entry in the notebook's sidebar minimap to the right.
- **Bold and lists make instructions easy to skim.** Students scan before they read closely.
- **Images make abstract ideas concrete.** A diagram, a photo, or a screenshot of expected output can greatly enhance a notebook.

## The formatting toolbar

Once a markdown cell is open for editing, a row of buttons appears above the text box:

- **H1 / H2 / H3** — turn the current line into a heading
- **Bold** / *Italic* — wrap the selected text
- **Link** — wrap the selection as a link
- **Bulleted list** / **Numbered list** — turn the selected lines into a list
- **Insert Image** — add an image from the notebook's attachments

Try editing this cell, creating new headings and changing the formatting of the text.

## Headings

A heading starts with one or more `#` characters, followed by a space:

```
# Biggest heading
## Medium heading
### Smaller heading
```

**Level-1 headings are special.** Codetto uses every `#` heading to build the notebook's outline in the sidebar minimap, so reserve `#` for the start of each new major topic. Use `##` or `###` for smaller sub-points within a topic — they show in the minimap too, just not as a collapsible section.

## Bold and italic

Wrap text in double asterisks for **bold**, or single asterisks for *italic*:

```
This is **bold** and this is *italic*.
```

Using the toolbar: select a word or phrase, then click **Bold** or *Italic*. Select already-bold text and click **Bold** again to remove the formatting.

## Lists

A line starting with `-` becomes a bulleted list item; a line starting with `1.` becomes a numbered list:

```
- First point
- Second point

1. Step one
2. Step two
```

The list buttons in the toolbar act on every line you've selected at once — select three lines of a paragraph and click **Bulleted list** to turn all three into one list.

## Links

Wrap the link text in square brackets, followed by the URL in parentheses:

```
[Python documentation](https://docs.python.org)
```

Using the toolbar: select the words that should become the link text, click **Link**, then just type over the placeholder URL that gets selected for you.

# Adding images

Markdown cells can also show images. This could be a diagram, a screenshot, or a photo. To add an image, it must already be **attached to the notebook** (paperclip icon in the top app bar) before it can be inserted here. See the separate Attachments guide if you haven't attached a file before.

## Using the Insert Image button

When editing a markdown cell, click **Insert Image** in the toolbar to open a small form:

- choose from the notebook's attached images
- add **alt text** — a short description, read aloud by screen readers
- pick an **alignment**: left, right, or center
- pick a **size**, as a percentage of the cell's width

Click Insert, and the image appears at the cursor.

## Sizing and aligning images

You can add a `{...}` right after the image markdown to control its size or alignment. For example:

```
{width=30 align=right}
```

`width` is a **percentage** of the cell's width, so the image stays proportional as the cell resizes. `align=right` (or `left`) floats the image so paragraph text wraps around it, like this:

![Codetto logo, floated to the right](/attachments/logo-light.png){width=30 align=right}

This paragraph exists just to show text wrapping around the floated image on the right. Once a paragraph runs past the bottom of the image, normal full-width text resumes underneath it, and the next paragraph starts clean below.

`align=center` centers the image instead, with no text wrap — a good choice for a standalone diagram or photo that should stand on its own:

![Codetto logo, centered](/attachments/logo-light.png){width=35 align=center}

# Tips for writing markdown

- Reserve level-1 headings (`#`) for the start of each major section. They become the collapsible entries a student sees in the sidebar minimap.
- **Bold** the first mention of a key term so it stands out on a skim.
- Keep image files reasonably small. Attached images share the same 5 MB per file / 20 MB per notebook limit as any other attachment.
- Always add short, meaningful alt text to an image. It's read aloud by screen readers and shown if the image ever fails to load.
