# Creating a 3D Scene

A 3D **scene** is a small world that you build with code. You decide what the sky looks like, lay down some ground, fill the world with shapes, and choose where the camera stands to look at it all.

Everything in this notebook comes from Codetto's `graphics3d` library. Positions are written as three numbers, `(x, y, z)`. If those are new to you, open [[4CBAB445-3D90-4AB1-AC79-0CBB84E70606|3D Coordinate System]] first. The short version: **x** is left and right, **y** is up and down, and **z** is forward and backward.

Run the cell below to make your first scene.

> **Tip:** Drag with the mouse (or a finger) to look around the scene. Scroll or pinch to zoom in and out.

```python
from codetto import graphics3d

scene = graphics3d.Scene()
scene.set_sky('#87ceeb')
scene.set_ground(length=10, width=10)
```

# Scene, Sky, and Ground

Those three lines are the start of nearly every 3D program you will write.

- `graphics3d.Scene()` creates the world and shows it below the cell. Keep it in a variable (here `scene`) because everything else is added to it.
- `scene.set_sky(color)` paints the sky. Colors are written as **hex codes** such as `'#87ceeb'`, the same codes web pages use.
- `scene.set_ground(length, width)` lays a flat floor, centered on the middle of the world. The top of the ground is at height `y = 0`.

The world is empty for now. Time to put something in it.

# Adding a Shape

Adding a shape always takes the same three steps:

1. **Create** it, for example `graphics3d.Shapes.Box(width=1, height=1, depth=1)`.
2. **Set it up**: where it goes, what color it is.
3. **Add** it to the scene with `scene.add(shape)`. A shape you forget to add never appears.

A shape's position is the position of its **center**. A box that is `1` tall and placed at `y = 0` would be half buried in the ground, so the box below sits at `y = 0.5`, half of its height.

Run the cell, then try changing the `0.5` to `0` and to `3`.

```python
scene = graphics3d.Scene()
scene.set_sky('#87ceeb')
scene.set_ground(length=10, width=10)

box = graphics3d.Shapes.Box(width=1, height=1, depth=1)
box.set_position(0, 0.5, 0)
box.set_color('#e94560')
scene.add(box)
```

# The Shape Collection

`graphics3d.Shapes` has seven shapes to build with. Each one has its own measurements:

| Shape | Measurements |
|---|---|
| `Box` | `width`, `height`, `depth` |
| `Sphere` | `diameter` |
| `Cylinder` | `diameter`, `height` |
| `Cone` | `diameter`, `height` |
| `Torus` (a ring) | `diameter`, `thickness` |
| `Plane` (a flat sheet) | `width`, `height` |
| `Text` | the text, then `size` and `depth` |

Sizes are in the same units as positions, so a box with `width=2` is exactly as wide as the gap between `x = 0` and `x = 2`.

The cell below lines up one of each. Drag the scene around to see them from every side.

```python
scene = graphics3d.Scene()
scene.set_sky('#1a1a2e')
scene.set_ground(length=16, width=16)

box = graphics3d.Shapes.Box(width=1.4, height=1.4, depth=1.4)
box.set_position(-6, 0.7, 0)
box.set_color('#e94560')
scene.add(box)

sphere = graphics3d.Shapes.Sphere(diameter=1.6)
sphere.set_position(-4, 0.8, 0)
sphere.set_color('#f5a623')
scene.add(sphere)

cylinder = graphics3d.Shapes.Cylinder(diameter=1.2, height=1.8)
cylinder.set_position(-2, 0.9, 0)
cylinder.set_color('#f8e71c')
scene.add(cylinder)

cone = graphics3d.Shapes.Cone(diameter=1.5, height=1.8)
cone.set_position(0, 0.9, 0)
cone.set_color('#44cc88')
scene.add(cone)

torus = graphics3d.Shapes.Torus(diameter=1.4, thickness=0.4)
torus.set_position(2, 0.2, 0)
torus.set_color('#4a90d9')
scene.add(torus)

plane = graphics3d.Shapes.Plane(width=1.4, height=1.8)
plane.set_position(4, 0.9, 0)
plane.set_color('#c77dff')
scene.add(plane)

text = graphics3d.Shapes.Text('3D', size=1.5, depth=0.4)
text.set_position(6.2, 0.1, 0)
text.set_color('#ff7eb6')
scene.add(text)

# Stand the camera back so the whole row fits (more on this later)
scene.camera.set_position(0, 4, -15)
```

# Position, Rotation, and Scale

Every shape understands the same three commands for moving and reshaping it:

- `set_position(x, y, z)` moves it.
- `set_rotation(x, y, z)` turns it. Each number is an angle in **degrees** around that axis, so `set_rotation(y=45)` spins the shape an eighth of a turn around the upright y axis, like a spinning top.
- `set_scale(x, y, z)` stretches it. `1` leaves that direction alone, `2` doubles it, and `0.5` halves it.

You can name just the directions you care about. `set_scale(y=3)` makes a shape three times taller and leaves its width and depth as they were.

The cell below has sliders for all three. Move a slider, then press **Run** again to rebuild the scene.

```python
SPIN = 30 #@param {type:"slider", min:0, max:180, step:15}
TILT = 0 #@param {type:"slider", min:0, max:90, step:15}
STRETCH = 1 #@param {type:"slider", min:0.5, max:3, step:0.5}
LIFT = 1 #@param {type:"slider", min:0, max:4, step:0.5}

scene = graphics3d.Scene()
scene.set_sky('#1a1a2e')
scene.set_ground(length=10, width=10)

box = graphics3d.Shapes.Box(width=2, height=1, depth=1)
box.set_position(0, LIFT, 0)
box.set_rotation(x=TILT, y=SPIN)
box.set_scale(y=STRETCH)
box.set_color('#f5a623')
scene.add(box)

print(f'Spun {SPIN} degrees, tilted {TILT} degrees, {STRETCH} times as tall.')
```

# Color and Transparency

`set_color` takes a hex code: a `#` followed by six characters. The first pair says how much **red** to mix in, the second pair how much **green**, and the third how much **blue**. Each pair runs from `00` (none) to `ff` (full).

| Hex code | Color |
|---|---|
| `'#ff0000'` | Pure red |
| `'#00ff00'` | Pure green |
| `'#0000ff'` | Pure blue |
| `'#ffffff'` | White (everything at full) |
| `'#000000'` | Black (nothing at all) |

The ground is a shape too. `set_ground` hands it back to you, so you can color it like anything else.

`set_alpha` makes a shape see-through: `1` is solid, `0` is invisible, and anything in between looks like tinted glass.

```python
scene = graphics3d.Scene()
scene.set_sky('#fdf6e3')

ground = scene.set_ground(length=10, width=10)
ground.set_color('#2a9d8f')

ball = graphics3d.Shapes.Sphere(diameter=1.2)
ball.set_position(0, 0.6, 0)
ball.set_color('#e76f51')
scene.add(ball)

glass = graphics3d.Shapes.Box(width=2, height=2, depth=2)
glass.set_position(0, 1, 0)
glass.set_color('#a8dadc')
glass.set_alpha(0.4)
scene.add(glass)
```

# Materials

A flat color only goes so far. A **material** makes a surface look like something real: brick, marble, grass, wood. Use `set_material` in place of `set_color`.

Materials are sorted into groups under `graphics3d.Material`. Here are some of them:

| Group | Materials |
|---|---|
| `Material.Bricks` | `DarkClay`, `RoughStone` |
| `Material.Marble` | `Brown`, `Gray`, `Black`, `Charcoal` |
| `Material.Grass` | `Bright`, `Dark`, `Olive` |
| `Material.Wood` | `Oak` |
| `Material.Tiles` | `Checkerboard`, `GreenMosaic`, `WoodHexagon`, `LimeGreen` |
| `Material.Planets` | `Earth`, `Moon`, `Mars`, `Jupiter`, `Saturn`, and more |
| `Material.Sports` | `Soccerball`, `Tennis` |

Type `graphics3d.Material.` in a code cell and the editor will list every group for you.

Two more commands fine-tune a material:

- `set_tiling(n)` repeats the picture `n` times across the surface. Big surfaces like the ground need this, otherwise one small picture is stretched across the whole thing and looks blurry.
- `set_glossiness(value)` sets the shine, from `0` (dull) to `1` (mirror-like). It only works on shapes that have a material.

```python
scene = graphics3d.Scene()
scene.set_sky('#87ceeb')

ground = scene.set_ground(length=20, width=20)
ground.set_material(graphics3d.Material.Grass.Bright)
ground.set_tiling(8)

wall = graphics3d.Shapes.Box(width=3, height=2, depth=0.5)
wall.set_position(-3, 1, 0)
wall.set_material(graphics3d.Material.Bricks.DarkClay)
scene.add(wall)

statue = graphics3d.Shapes.Sphere(diameter=2, segments=32)
statue.set_position(0.5, 1, 0)
statue.set_material(graphics3d.Material.Marble.Gray)
statue.set_glossiness(0.9)
scene.add(statue)

ball = graphics3d.Shapes.Sphere(diameter=1, segments=32)
ball.set_position(3, 0.5, 0)
ball.set_material(graphics3d.Material.Sports.Soccerball)
scene.add(ball)
```

# A Sky to Match

The sky can be more than one color. `graphics3d.Sky` has five ready-made skies that wrap all the way around your scene:

| Sky | What you see |
|---|---|
| `Sky.CLOUDS` | Blue sky with fluffy clouds |
| `Sky.PURE_SKY` | A clear day |
| `Sky.DEEP_SPACE` | Stars in every direction |
| `Sky.MODERN_BUILDINGS` | A city square |
| `Sky.ORLANDO_STADIUM` | The inside of a stadium |

These skies also change how shiny materials look, because glossy surfaces **reflect** the sky around them. Look closely at the marble ball below, then swap `Sky.CLOUDS` for another sky and run the cell again.

```python
scene = graphics3d.Scene()
scene.set_sky(graphics3d.Sky.CLOUDS)

ground = scene.set_ground(length=20, width=20)
ground.set_material(graphics3d.Material.Tiles.Checkerboard)
ground.set_tiling(6)

ball = graphics3d.Shapes.Sphere(diameter=3, segments=48)
ball.set_position(0, 1.5, 0)
ball.set_material(graphics3d.Material.Marble.Black)
ball.set_glossiness(1)
scene.add(ball)
```

# Your Own Pictures

If none of the built-in materials fit, `set_texture` wraps any picture around a shape. Give it the path to an image file.

The cell below uses one of the sample pictures that ships with Codetto. To use a picture of your own, add it with the paperclip button at the top of the notebook, then use its path, for example `'/attachments/my_photo.jpg'`.

```python
from codetto import samples

scene = graphics3d.Scene()
scene.set_sky('#fdf6e3')
scene.set_ground(length=10, width=10)

crate = graphics3d.Shapes.Box(width=2, height=2, depth=2)
crate.set_position(0, 1, 0)
crate.set_rotation(y=30)
crate.set_texture(samples.Images.Cat)
scene.add(crate)
```

# Placing the Camera

So far you have moved the view by dragging. Your code can place the camera too, through `scene.camera`. The camera lives in the same world as the shapes and uses the same `(x, y, z)` positions.

- `scene.camera.set_position(x, y, z)` puts the camera somewhere.
- `scene.camera.look_at(x, y, z)` turns it to face a point. You can also pass a shape: `look_at(box)`.
- `scene.camera.set_distance(d)` moves it closer to, or further from, whatever it is looking at.

A new scene starts with the camera at about `(0, 8.5, -8.5)`: above the ground and pulled back toward you, looking at the center of the world.

| Camera position | The view |
|---|---|
| `(0, 14, 0)` | Straight down from above, like a map |
| `(0, 1, -10)` | From ground level, looking across |
| `(12, 3, 0)` | From the right-hand side |

Camera commands can be joined together on one line, as the last line of the cell below does. Try the sliders, and remember that you can still drag the view afterwards.

```python
CAMERA_HEIGHT = 2 #@param {type:"slider", min:1, max:16, step:1}
CAMERA_DISTANCE = 12 #@param {type:"slider", min:4, max:30, step:2}

scene = graphics3d.Scene()
scene.set_sky(graphics3d.Sky.PURE_SKY)

ground = scene.set_ground(length=20, width=20)
ground.set_material(graphics3d.Material.Grass.Dark)
ground.set_tiling(8)

tower = graphics3d.Shapes.Cylinder(diameter=1.5, height=5)
tower.set_position(0, 2.5, 0)
tower.set_material(graphics3d.Material.Bricks.RoughStone)
scene.add(tower)

roof = graphics3d.Shapes.Cone(diameter=2.2, height=1.5)
roof.set_position(0, 5.75, 0)
roof.set_color('#b23a48')
scene.add(roof)

scene.camera.set_position(0, CAMERA_HEIGHT, -CAMERA_DISTANCE).look_at(tower)
```

# Build Your Own Scene

Here is a small scene that uses everything from this notebook: a sky, a textured ground, shapes with colors and materials, a rotation, and a camera position. The trees are made by a `for` loop that builds the same two shapes at four different spots.

Make it yours. Some ideas:

- Give the house a door, a chimney, or a second floor.
- Add more trees by adding positions to the list.
- Change the season: `Material.Snow.Fresh` on the ground, and a different sky.
- Put a name sign in front of the house with `Shapes.Text`.
- Find a camera position that shows off your scene at its best.

When you are ready to make things move, continue with [[B10CBE41-6C45-4FFA-9D4A-1026925C0130|Lights, Camera, Animation]].

```python
scene = graphics3d.Scene()
scene.set_sky(graphics3d.Sky.PURE_SKY)

ground = scene.set_ground(length=24, width=24)
ground.set_material(graphics3d.Material.Grass.Bright)
ground.set_tiling(10)

# The house: a brick box with a four-sided cone for a roof
house = graphics3d.Shapes.Box(width=3, height=2, depth=3)
house.set_position(0, 1, 0)
house.set_material(graphics3d.Material.Bricks.DarkClay)
scene.add(house)

roof = graphics3d.Shapes.Cone(diameter=5, height=1.6, tessellation=4)
roof.set_position(0, 2.8, 0)
roof.set_rotation(y=45)
roof.set_color('#7b2d26')
scene.add(roof)

# The trees: a trunk and a ball of leaves at each position
for x, z in [(-5, 2), (-4, -3), (5, 1), (4, -4)]:
  trunk = graphics3d.Shapes.Cylinder(diameter=0.4, height=1.6)
  trunk.set_position(x, 0.8, z)
  trunk.set_material(graphics3d.Material.Wood.Oak)
  scene.add(trunk)

  leaves = graphics3d.Shapes.Sphere(diameter=2)
  leaves.set_position(x, 2.3, z)
  leaves.set_color('#2d6a4f')
  scene.add(leaves)

scene.camera.set_position(3, 5, -14).look_at(house)
```
