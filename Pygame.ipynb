# Making Games with Pygame

**Pygame** is one of the most popular Python libraries for making 2D games. It gives you a screen to draw on, a way to read the keyboard and mouse, and tools for animation and sound.

In this notebook you'll learn the building blocks that every Pygame game is made from, one at a time. By the end you'll have a small game where you chase coins around the screen.

# Creating a Screen

Every Pygame program starts the same way. `pygame.init()` switches the library on, and `pygame.display.set_mode((width, height))` creates the **screen**, the surface that everything is drawn on.

Colors are written as three numbers from 0 to 255: how much **red, green, and blue** to mix. `(255, 0, 0)` is pure red, `(0, 0, 0)` is black, and `(255, 255, 255)` is white.

Drawing happens in memory first. Nothing appears until you call `pygame.display.flip()`, which shows the finished picture all at once.

````python
import pygame

pygame.init()

WIDTH, HEIGHT = 400, 300
screen = pygame.display.set_mode((WIDTH, HEIGHT))

screen.fill((25, 30, 60))
pygame.display.flip()
````

# Drawing Shapes

The `pygame.draw` functions add shapes to the screen. Each one takes the screen, a color, and where to draw.

Positions are measured in pixels from the **top-left corner**. `x` grows to the right and `y` grows *downward*, so `(0, 0)` is the top-left and `(400, 300)` is the bottom-right of our screen.

| Function | What you give it |
|---|---|
| `pygame.draw.rect` | `(x, y, width, height)` |
| `pygame.draw.circle` | center `(x, y)` and a radius |
| `pygame.draw.line` | a start point, an end point, and a thickness |
| `pygame.draw.polygon` | a list of corner points |

Shapes are drawn in order, so later shapes cover earlier ones. That is why the house below hides part of the horizon line.

````python
screen = pygame.display.set_mode((WIDTH, HEIGHT))
screen.fill((135, 206, 235))  # sky

pygame.draw.rect(screen, (70, 160, 80), (0, 220, WIDTH, 80))      # grass
pygame.draw.line(screen, (255, 255, 255), (0, 220), (WIDTH, 220), 2)  # horizon
pygame.draw.circle(screen, (255, 215, 0), (330, 60), 30)          # sun
pygame.draw.rect(screen, (220, 170, 120), (140, 150, 120, 90))    # house
pygame.draw.polygon(screen, (160, 60, 50),
                    [(125, 150), (200, 90), (275, 150)])          # roof
pygame.draw.rect(screen, (90, 60, 40), (185, 190, 30, 50))        # door

pygame.display.flip()
````

## Drawing with a Loop

Because drawing is just Python code, you can use a loop to draw many shapes at once. Pygame also understands color names like `"gold"` and `"tomato"`.

Drag the slider and change the color to redraw the row of circles.

````python
COUNT = 6 #@param {type:"slider", min:1, max:12, step:1}
COLOR = "gold" #@param ["gold", "tomato", "skyblue", "limegreen"]

screen = pygame.display.set_mode((WIDTH, HEIGHT))
screen.fill((25, 30, 60))

gap = WIDTH / COUNT
for i in range(COUNT):
  x = gap * i + gap / 2
  pygame.draw.circle(screen, COLOR, (x, HEIGHT / 2), gap * 0.4)

pygame.display.flip()
````

# Adding Text

Text takes two steps. First a **font** turns your words into a small picture using `font.render(text, True, color)`. Then `screen.blit(picture, position)` copies that picture onto the screen. **Blit** is the Pygame word for "copy one picture onto another".

The `True` in `render` asks for smooth edges on the letters.

````python
screen = pygame.display.set_mode((WIDTH, HEIGHT))
screen.fill((25, 30, 60))

big_font = pygame.font.SysFont(None, 48)
small_font = pygame.font.SysFont(None, 24)

title = big_font.render("Hello, Pygame!", True, (255, 215, 0))
subtitle = small_font.render("Score: 0", True, (255, 255, 255))

screen.blit(title, (80, 120))
screen.blit(subtitle, (10, 10))
pygame.display.flip()
````

# The Game Loop

So far every picture has stood still. To make things move, a game redraws the screen many times a second, changing it a little each time. This is the **game loop**, and each pass through it is called a **frame**.

Every frame does the same four jobs:

1. **Handle events**, the things that just happened, such as a key press
2. **Update** the positions of everything in the game
3. **Draw** the screen again from scratch
4. **Flip** the picture onto the display, then wait so the game runs at a steady speed

`clock.tick(60)` does the waiting. It holds the loop at 60 frames per second on every computer.

The ball below moves by `dx` and `dy` pixels each frame. Bouncing is just flipping the sign of one of them. Press **Escape** to end the loop, or use the Stop button.

````python
screen = pygame.display.set_mode((WIDTH, HEIGHT))
clock = pygame.time.Clock()

x, y = 200, 150
dx, dy = 3, 2

running = True
while running:
  # 1. Handle events
  for event in pygame.event.get():
    if event.type == pygame.KEYDOWN and event.key == pygame.K_ESCAPE:
      running = False

  # 2. Update
  x += dx
  y += dy
  if x < 15 or x > WIDTH - 15:
    dx = -dx
  if y < 15 or y > HEIGHT - 15:
    dy = -dy

  # 3. Draw
  screen.fill((25, 30, 60))
  pygame.draw.circle(screen, (255, 215, 0), (x, y), 15)

  # 4. Flip and wait
  pygame.display.flip()
  clock.tick(60)
````

# Keyboard Input

There are two ways to read the keyboard, and games use both.

- **Events** tell you the moment a key goes down. A `KEYDOWN` event happens *once* per press, which suits actions like jumping or firing.
- `pygame.key.get_pressed()` tells you which keys are being **held** right now. Check it every frame for smooth movement.

Run the cell, then hold the arrow keys to move the square. Tap **Space** to change its color, and press **Escape** to finish.

````python
screen = pygame.display.set_mode((WIDTH, HEIGHT))
clock = pygame.time.Clock()

x, y = 180, 130
colors = ["tomato", "gold", "skyblue", "limegreen"]
color = 0

running = True
while running:
  for event in pygame.event.get():
    if event.type == pygame.KEYDOWN:
      if event.key == pygame.K_ESCAPE:
        running = False
      if event.key == pygame.K_SPACE:
        color = (color + 1) % len(colors)

  keys = pygame.key.get_pressed()
  if keys[pygame.K_LEFT]:
    x -= 4
  if keys[pygame.K_RIGHT]:
    x += 4
  if keys[pygame.K_UP]:
    y -= 4
  if keys[pygame.K_DOWN]:
    y += 4

  screen.fill((25, 30, 60))
  pygame.draw.rect(screen, colors[color], (x, y, 40, 40))
  pygame.display.flip()
  clock.tick(60)
````

# Mouse Input

The mouse works the same way. A `MOUSEBUTTONDOWN` event happens once per click, `pygame.mouse.get_pos()` gives the pointer's `(x, y)` position, and `pygame.mouse.get_pressed()` tells you which buttons are held down.

This cell is a tiny paint program. Hold the mouse button and drag across the screen to paint. Each new click picks a new color.

Notice that the screen is filled only *once*, before the loop starts. Because nothing clears it each frame, the paint stays where you put it.

````python
import random

screen = pygame.display.set_mode((WIDTH, HEIGHT))
clock = pygame.time.Clock()
screen.fill((25, 30, 60))
color = "gold"

running = True
while running:
  for event in pygame.event.get():
    if event.type == pygame.KEYDOWN and event.key == pygame.K_ESCAPE:
      running = False
    if event.type == pygame.MOUSEBUTTONDOWN:
      color = random.choice(["gold", "tomato", "skyblue", "limegreen"])

  if pygame.mouse.get_pressed()[0]:
    pygame.draw.circle(screen, color, pygame.mouse.get_pos(), 10)

  pygame.display.flip()
  clock.tick(60)
````

# Rectangles and Collisions

Most games need to know when two things touch. Pygame's **`Rect`** stores a position and a size, and comes with handy tools:

- `rect.x` and `rect.y` move it around
- `rect.colliderect(other)` is `True` when two rectangles overlap
- `rect.clamp_ip(area)` pushes it back inside an area, so it can't leave the screen

Here is a complete mini game. Use the arrow keys to steer the square into the coin. Each time they touch, you score a point and the coin jumps somewhere new.

````python
import random

screen = pygame.display.set_mode((WIDTH, HEIGHT))
clock = pygame.time.Clock()
font = pygame.font.SysFont(None, 28)

player = pygame.Rect(180, 130, 40, 40)
coin = pygame.Rect(300, 60, 20, 20)
score = 0

running = True
while running:
  for event in pygame.event.get():
    if event.type == pygame.KEYDOWN and event.key == pygame.K_ESCAPE:
      running = False

  keys = pygame.key.get_pressed()
  if keys[pygame.K_LEFT]:
    player.x -= 4
  if keys[pygame.K_RIGHT]:
    player.x += 4
  if keys[pygame.K_UP]:
    player.y -= 4
  if keys[pygame.K_DOWN]:
    player.y += 4
  player.clamp_ip(screen.get_rect())

  if player.colliderect(coin):
    score += 1
    coin.x = random.randint(0, WIDTH - coin.width)
    coin.y = random.randint(0, HEIGHT - coin.height)

  screen.fill((25, 30, 60))
  pygame.draw.rect(screen, "tomato", player)
  pygame.draw.ellipse(screen, "gold", coin)
  screen.blit(font.render(f"Score: {score}", True, "white"), (10, 10))
  pygame.display.flip()
  clock.tick(60)
````

# Sound Effects

A game feels much more alive with sound. `pygame.mixer.Sound(path)` loads a sound effect once, and calling `.play()` on it plays it without pausing the game.

Codetto comes with a set of sample sounds you can use, found in `samples.Sounds`.

````python
from codetto import samples

coin_sound = pygame.mixer.Sound(samples.Sounds.CoinPickup)
coin_sound.play()
````
