# Create and Embed Diagrams

**Diagram cells** enable you to embed diagrams in notebooks. This could be a flowchart, a sequence diagram, a class diagram, and more - using the open-source **Mermaid** syntax. No drawing required: describe the boxes and arrows, and Codetto renders it for you.

## Why use a diagram cell?

- **Easy to update.** Editing a text description is faster than re-drawing boxes and arrows by hand.
- **Consistent look.** Every diagram uses the same clean Codetto style automatically.
- **Great for process and structure.** Show the steps of an algorithm, the flow of data through a program, or how classes relate to each other, right next to the code that implements it.

## The source format

A diagram cell's source is plain Mermaid text. There's nothing else to learn beyond Mermaid's own syntax. Every diagram starts by naming its type:

```
flowchart TD
    A[Start] --> B[End]
```

`TD` means top-down; other flowcharts can run `LR` (left-to-right), `BT`, or `RL`. Each line after that adds a box (in square brackets) or an arrow between two boxes.

## Diagram types

Mermaid supports many diagram types beyond flowcharts:

- `flowchart` — boxes and arrows, for a process or decision tree
- `sequenceDiagram` — messages passing between participants over time
- `classDiagram` — classes, their fields/methods, and relationships
- `stateDiagram-v2` — states and the transitions between them
- `erDiagram` — entity-relationship diagrams for database design
- `gantt` — a project timeline
- `pie` — a simple pie chart

## Colors and themes

Every diagram uses Codetto's own green color scheme by default. To use different colors for one specific diagram, add an `%%{init: ...}%%` directive as the very first line:

```
%%{init: {"theme": "base", "themeVariables": {"primaryColor": "#1e88e5"}}}%%
flowchart TD
    A[Start] --> B[End]
```

This is standard Mermaid syntax, not a Codetto-specific feature — anything documented for Mermaid's `init` directive works here.

## Viewing and saving

Hover over a diagram to reveal a magnify icon. You can click it to open a larger view in a dialog, with a **Save** button that downloads the diagram as an SVG file, useful for pasting into a slide deck or document.
