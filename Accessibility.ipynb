# Accessibility

Students use Codetto in different ways. Some use a keyboard and no mouse. Some use a **screen reader**, which reads the page aloud. Some need larger text, or find movement on the screen distracting. This guide shows what Codetto does to support these students, and the small things you can do in your own notebooks to help.

## Getting around with the keyboard

Everything in Codetto can be reached without a mouse. Try it now: put the mouse aside and use the keys below.

- **Tab** moves forward to the next button, link, or field. **Shift+Tab** moves back.
- **Enter** or **Space** presses the button you're on.
- **Arrow keys** move through the choices in a menu, a set of answers, or a row of tabs.
- **Escape** closes a menu or a dialog and puts you back where you were.
- **Page Up**, **Page Down**, and the arrow keys scroll the page.

A dark outline shows which item the keyboard is on. If you lose track of it, press Tab once and look for the outline.

Two shortcuts save a lot of pressing:

- **Skip to content.** The first press of Tab on any page shows a **Skip to content** button in the top-left corner. Press Enter to jump straight past the menu.
- **Search.** On the Home and Explore pages, press **Ctrl+K** (or **⌘K** on a Mac) to search for a notebook.

Inside a notebook, the **Notebook Map** (the panel on the right) is the quickest way to move around. Tab to a row and press Enter, and the keyboard moves to that part of the notebook.

## Using a code cell with the keyboard

Code cells need a little more explanation, because the Tab key has two jobs. Everywhere else, Tab moves to the next item. Inside the code editor, Tab indents a line, which is what Python needs.

So there is a different way to leave the editor:

- **To go in:** press Tab until the editor is outlined. You can start typing straight away.
- **To indent:** press Tab. **Shift+Tab** removes an indent.
- **To run the code:** press **Ctrl+Enter** (or **⌘+Enter** on a Mac). You stay in the editor, ready to make another change.
- **To leave:** press **Escape**, then **Tab**. You move on to the buttons below the editor. **Escape**, then **Shift+Tab** takes you back the way you came.

When you arrive in an editor using the keyboard, a reminder appears just below it: *Press Esc, then Tab to leave the editor.* Students who use a mouse never see it.

Try it with the cell below. Tab into the editor, change `4` to `6`, run the code with Ctrl+Enter, then press Escape and Tab to move on.

````python
#@title Count up
for number in range(1, 4):
  print(number)
````

It's worth showing the whole class **Escape, then Tab** in the first lesson. Students who use the keyboard need it in every notebook, and it is the one step that isn't obvious.

## Screen readers

A screen reader reads the page aloud and lets a student move through it by heading, by button, or line by line. 

Here is what a student using a screen reader hears:

- **Where they are.** Each page announces its name when it opens, including the title of a notebook.
- **What each button does.** Every button has a spoken name, such as "Run code" or "Clear output", even when it only shows an icon.
- **Which cell they're in.** Each code cell is announced as a group, such as "Code cell 3". A cell with a `#@title` is announced by its title.
- **What their code did.** Running a cell announces "Running code…", then the result. Short output is read out in full. Longer output is summed up as a number of lines, which the student can then read in the output area. If the code fails, the last line of the error is read out.
- **Questions and answers.** A Check for Understanding question is read together with its answer box or choices, and "Correct!" or "Not quite" is announced after submitting.

Run the cell above again and listen, if you have a screen reader switched on. On a Mac, press **⌘+F5** to turn VoiceOver on or off.

## Describing charts and drawings with #@alt

A chart or a drawing is a picture, and a screen reader can't tell what a picture shows. Without help, a student hears only "Image output".

A `#@alt` comment fixes this. Add it to the code cell that makes the picture, followed by a short description:

```
#@alt Bar chart of rainfall. April is highest at 80 mm.
```

Because `#@alt` is a comment, Python ignores it and the code runs exactly as before. Codetto uses the description as the spoken name of the cell's chart, drawing, or 3D scene.

A few things to know:

- **It can go on any line** of the cell. The top is easiest to find. If a cell has more than one `#@alt`, the first is used.
- **Say what the picture shows, then what matters about it.** "Line chart of temperature" is a start. "Line chart of temperature, rising steadily from 12°C in January to 28°C in July" lets the student answer the same questions as everyone else.
- **Keep it to a sentence or two.**
- **Update it when the code changes.** A description of last week's chart is worse than none.

Try the example below. Run it to see the chart. A student using a screen reader hears the description in place of the picture.

````python
#@title Rainfall chart
#@alt Bar chart of rainfall over four months. April is highest at 80 mm and June is lowest at 20 mm.
import matplotlib.pyplot as plt

months = ["March", "April", "May", "June"]
rainfall = [50, 80, 40, 20]

plt.bar(months, rainfall)
plt.ylabel("Rainfall (mm)")
plt.show()
````

Writing a `#@alt` line also makes a good exercise for the whole class. Describing a chart in one sentence is a quick check that a student can read it.

Pictures in markdown cells have descriptions too. The text in square brackets is what a screen reader reads, so `![A micro:bit showing a heart](/attachments/heart.jpg)` is far more useful than `![image](/attachments/heart.jpg)`. ## Text, color, and contrast

Text in Codetto is dark on a light background, with enough difference between the two to stay readable for students with low vision, on a dim screen, or on a projector in a bright room. The code editor's colors were chosen to stay distinct for students who are colorblind.

Color is never the only signal. A button that is switched on is filled in, as well as green. A correct answer shows a tick and the word "Correct!", as well as a green tint.

The same two ideas are worth following in what you make:

- **Keep text and background well apart.** When you choose colors for a chart or a slide, dark on light or light on dark is easiest to read. Pale colors on white are the usual problem.
- **Don't rely on color alone.** In a chart with several lines, add labels or use different markers, so "the red line" isn't the only way to tell them apart.

## Making things bigger

On a computer, students can make the whole page bigger with the browser's own zoom: **Ctrl** and **+** (or **⌘** and **+** on a Mac). **Ctrl** and **0** returns to normal size. The page rearranges itself to fit, so there's no need to scroll sideways.

On a tablet or phone, zooming with two fingers is **switched off to begin with**. Students often zoom by accident while tapping and scrolling, and then struggle to get back.

A student who needs to zoom can switch it on:

1. Open **Settings** from the menu on the left.
2. Find the **Accessibility** section.
3. Switch on **Pinch to Zoom**.

The setting is remembered on that device, so it only needs doing once per tablet. Drawing areas, games, and 3D scenes keep their own touch controls and don't zoom, whichever way the setting is.

## Reducing movement

Some students find movement on the screen distracting or uncomfortable. Computers, tablets, and phones have a setting for this, usually called **Reduce motion** and found in the device's accessibility settings.

When it's switched on, Codetto follows it. Animations stop, the page jumps to a new position without sliding, and the confetti after a correct answer is left out. There's nothing to change in Codetto itself.

## Captions for videos

A video with speech needs **captions** for students who are deaf or hard of hearing, and they help anyone watching in a noisy room. A video cell can carry captions with one extra setting:

```
{
  "url": "https://example.com/intro.mp4",
  "title": "Introduction",
  "captions": "https://example.com/intro.vtt"
}
```

The captions are a separate text file in the **WebVTT** format, which usually ends in `.vtt`. Most video editing and captioning tools can save one. The file can be a link, as above, or a file attached to the notebook, such as `"/attachments/intro.vtt"`.

Students switch captions on from the video's own controls. A video set to play automatically always shows its controls, so it can be paused.

## A checklist for your own notebooks

Codetto takes care of the buttons, the keyboard, and the colors. These habits take care of the rest:

- **Use headings in order.** Start with one title (`#`), then sections (`##`), then smaller parts (`###`). Students using a screen reader jump between headings, and the Notebook Map is built from them.
- **Name your code cells.** A `#@title` on the first line turns "Code cell 7" into something a student can find. See [[91A04016-71D4-4559-9154-964F11B0C449]].
- **Describe every picture.** Use the square brackets for images in markdown, and `#@alt` for charts and drawings made by code.
- **Add captions to videos with speech.**
- **Write links that make sense alone.** "Read the micro:bit guide" tells a student where a link goes. "Click here" doesn't.
- **Try it without the mouse.** Before a lesson, tab through your notebook once. If you get stuck, so will a student.

## Activities to plan for

Some activities depend on one way of working, and no setting can change that:

- **Drawing and clicking.** Drawing on a canvas with the mouse or a finger, or clicking on shapes in a 3D scene, has no keyboard equivalent.
- **The webcam.** Activities that track a face, a hand, or a pose need a student to be seen by the camera.
- **Sound.** Playing notes, recording a voice, and text to speech depend on hearing or speaking.
- **Charts made by students.** A student can't describe a chart they can't see. Ask them to print the numbers behind it as well.

None of this is a reason to avoid these activities. When one of them is in your plan, think about who is in the room, and have another way to reach the same goal: a partner to work with, a version that uses the keyboard, or the same data as text.
