# Drawing on a Canvas

A **canvas** is a rectangle of pixels that you can draw on with code. Games, charts, paint programs, and map apps all draw their pictures onto a canvas like this one.

Every point on the canvas has an address made of two numbers: `x` (how far across, from the left) and `y` (how far down, from the top). The point `(0, 0)` is the **top-left** corner, and `y` grows *downward*, which is the opposite of the graphs you draw in maths.

You make a canvas with `graphics.canvas()`. Run the cell below to create one and paint it a solid colour.

```python
from codetto import graphics

canvas = graphics.canvas(400, 300)
ctx = canvas.get_context('2d')

ctx.fill_style = '#ffe0b2'
ctx.fill_rect(0, 0, 400, 300)

print('Canvas ready.')
```

# The Drawing Context

You never draw on the canvas directly. Instead you ask it for a **drawing context** with `canvas.get_context('2d')`, and every drawing command goes through that.

The context works in two ways. You *set a property* to choose how the next shape will look, then you *call a method* to actually draw it:

- `ctx.fill_style = '#e74c3c'` chooses the fill colour (any CSS colour name or hex code).
- `ctx.fill_rect(x, y, width, height)` draws a filled rectangle with the top-left corner at `(x, y)`.

A colour you set stays in effect until you change it, so you can draw several shapes in the same colour without repeating yourself.

```python
canvas = graphics.canvas(400, 200)
ctx = canvas.get_context('2d')

ctx.fill_style = '#4a90d9'
ctx.fill_rect(20, 20, 100, 100)
ctx.fill_rect(140, 20, 100, 100)

ctx.fill_style = '#e67e22'
ctx.fill_rect(260, 20, 100, 100)

print('Three squares, two colours.')
```

# Rectangles and Outlines

A filled shape is solid all the way through. An **outline** (also called a stroke) draws only the edge and leaves the middle empty.

- `ctx.stroke_style` sets the outline colour, and `ctx.line_width` sets how thick it is.
- `ctx.stroke_rect(x, y, width, height)` draws a rectangle outline.
- `ctx.clear_rect(x, y, width, height)` does the opposite of drawing: it erases a rectangular area back to transparent, like a rubber.

The code below draws a filled square, an outlined square next to it, and then erases a hole in the filled one.

```python
canvas = graphics.canvas(400, 200)
ctx = canvas.get_context('2d')

ctx.fill_style = '#27ae60'
ctx.fill_rect(30, 40, 120, 120)

ctx.stroke_style = '#8e44ad'
ctx.line_width = 6
ctx.stroke_rect(230, 40, 120, 120)

ctx.clear_rect(60, 70, 60, 60)

print('Filled, outlined, and erased.')
```

# Lines and Paths

To draw a straight line you build a **path**: a route made of points that the pen moves along.

1. `ctx.begin_path()` starts a fresh path.
2. `ctx.move_to(x, y)` lifts the pen and puts it down at a starting point.
3. `ctx.line_to(x, y)` draws a line from where the pen is to a new point. You can call it many times to make a connected shape.
4. `ctx.stroke()` finally inks the whole path using the current `stroke_style` and `line_width`.

Nothing appears until you call `stroke()` at the end.

```python
canvas = graphics.canvas(400, 240)
ctx = canvas.get_context('2d')

ctx.stroke_style = '#c0392b'
ctx.line_width = 4

ctx.begin_path()
ctx.move_to(40, 200)
ctx.line_to(120, 40)
ctx.line_to(200, 200)
ctx.line_to(280, 40)
ctx.line_to(360, 200)
ctx.stroke()

print('A zig-zag drawn as one path.')
```

# Circles and Arcs

Circles are drawn with `ctx.arc(x, y, radius, start_angle, end_angle)`. The first two numbers are the centre, then the radius, then how far around to sweep.

The angles are measured in **radians**, not degrees. A full turn all the way around is `2 * math.pi` (about `6.28`), so a whole circle goes from `0` to `2 * math.pi`. A half turn is `math.pi`.

An arc is part of a path, so the pattern is the same as before: `begin_path()`, then `arc(...)`, then either `fill()` for a solid disc or `stroke()` for a ring.

```python
import math

canvas = graphics.canvas(400, 200)
ctx = canvas.get_context('2d')

ctx.fill_style = '#f1c40f'
ctx.begin_path()
ctx.arc(120, 100, 60, 0, 2 * math.pi)
ctx.fill()

ctx.stroke_style = '#2c3e50'
ctx.line_width = 5
ctx.begin_path()
ctx.arc(280, 100, 60, 0, math.pi)
ctx.stroke()

print('A full disc and a half-circle.')
```

# Writing Text

You can print words straight onto the canvas.

- `ctx.font` sets the size and typeface as one string, like `'28px sans-serif'`.
- `ctx.fill_text(message, x, y)` draws the text. Here `(x, y)` is the **left end of the baseline** the letters sit on, not the top-left corner, so give `y` some room from the top edge.

Text uses `fill_style`, the same colour property the filled shapes use.

```python
canvas = graphics.canvas(400, 160)
ctx = canvas.get_context('2d')

ctx.fill_style = '#16a085'
ctx.font = '40px sans-serif'
ctx.fill_text('Hello, canvas!', 20, 70)

ctx.fill_style = '#7f8c8d'
ctx.font = '18px sans-serif'
ctx.fill_text('Drawn with ctx.fill_text', 20, 110)

print('Text drawn.')
```

# Colour Gradients

A **gradient** is a smooth blend from one colour to another. Instead of giving `fill_style` a single colour, you give it a gradient object.

- `ctx.create_linear_gradient(x0, y0, x1, y1)` makes a gradient that runs along the line from the first point to the second.
- `gradient.add_color_stop(position, colour)` pins a colour somewhere along that line. `position` is between `0` (the start) and `1` (the end).

Once the gradient is built, set it as `ctx.fill_style` and draw a shape as usual.

```python
canvas = graphics.canvas(400, 200)
ctx = canvas.get_context('2d')

gradient = ctx.create_linear_gradient(0, 0, 400, 0)
gradient.add_color_stop(0, '#ff6b6b')
gradient.add_color_stop(0.5, '#feca57')
gradient.add_color_stop(1, '#48dbfb')

ctx.fill_style = gradient
ctx.fill_rect(0, 0, 400, 200)

print('A three-colour sunset.')
```

# Draw with Sliders

Because the drawing is just code, you can drive it with **form fields**. The cell below has a slider, a dropdown, and a checkbox. It draws a simple face from those values.

This cell re-runs on its own every time you change a control, so drag the slider and switch the smile on and off to watch the face change.

```python
import math

FACE_SIZE = 130 #@param {type:"slider", min:60, max:180, step:10}
FACE_COLOR = "#ffcc66" #@param ["#ffcc66", "#88cc88", "#88bbff", "#ff99bb"]
SMILE = True #@param {type:"boolean"}

canvas = graphics.canvas(300, 300)
ctx = canvas.get_context('2d')
cx, cy = 150, 150

ctx.fill_style = FACE_COLOR
ctx.begin_path()
ctx.arc(cx, cy, FACE_SIZE / 2, 0, 2 * math.pi)
ctx.fill()

ctx.fill_style = '#2c3e50'
for eye_x in (cx - 25, cx + 25):
  ctx.begin_path()
  ctx.arc(eye_x, cy - 20, 8, 0, 2 * math.pi)
  ctx.fill()

ctx.stroke_style = '#2c3e50'
ctx.line_width = 5
ctx.begin_path()
if SMILE:
  ctx.arc(cx, cy + 10, 35, 0.15 * math.pi, 0.85 * math.pi)
else:
  ctx.arc(cx, cy + 55, 35, 1.15 * math.pi, 1.85 * math.pi)
ctx.stroke()

print(f'Face size {FACE_SIZE}, smiling: {SMILE}')
```

# Drawing Pictures

You can also put real image files on the canvas!

The quickest way is `graphics.draw_image(canvas, path)`, which stretches the picture to fill the whole canvas. For more control you can `graphics.load_image(path)` first and then use the image to make a repeating **pattern** with `ctx.create_pattern(image, 'repeat')`, exactly like using a gradient as the fill.

This example fills the top half of the canvas with a stretched photo (one of the sample files that ships with Codetto) and the bottom half with a tiled pattern of the same photo.

```python
from codetto import samples

canvas = graphics.canvas(400, 300)
ctx = canvas.get_context('2d')

graphics.draw_image(canvas, samples.Images.City)

img = graphics.load_image(samples.Images.City)
pattern = ctx.create_pattern(img, 'repeat')
ctx.fill_style = pattern
ctx.fill_rect(0, 150, 400, 150)

print('One picture, two ways.')
```

You can also bring in your own photos and images. Attached to this notebook is a picture of a cat. (Click on the paperclip icon in the top right and look for `cat.jpg`).

We can use the following code to create a new canvas and display this image.

```python
canvas = graphics.canvas(400,265)
ctx = canvas.get_context('2d')

graphics.draw_image(canvas, "/attachments/cat.jpg")
```

# Reacting to Clicks and Keys

So far every drawing has been painted once and left alone. A canvas can also **listen** for the mouse and keyboard while it stays on screen.

- `canvas.on_click(function)` calls your function every time the canvas is clicked, passing it the `x` and `y` of the click.
- `canvas.on_key(key, function)` calls your function when that key is pressed. The key is a plain string, like `' '` for the space bar or `'c'` for the C key.
- `canvas.run()` keeps the cell alive so those handlers can fire. It only stops when your code calls something that ends it, or when you press the **Stop** button.

Run the cell, then click around the canvas and press the space bar. Press Stop when you are done.

```python
import math

canvas = graphics.canvas(400, 300)
ctx = canvas.get_context('2d')
ctx.fill_style = '#1b263b'
ctx.fill_rect(0, 0, 400, 300)

colors = ['#ff6b6b', '#feca57', '#1dd1a1', '#54a0ff', '#ff9ff3']

def draw_dot(x, y):
  ctx.fill_style = colors[(x + y) % len(colors)]
  ctx.begin_path()
  ctx.arc(x, y, 12, 0, 2 * math.pi)
  ctx.fill()

def clear_canvas():
  ctx.fill_style = '#1b263b'
  ctx.fill_rect(0, 0, 400, 300)
  print('cleared')

canvas.on_click(draw_dot)
canvas.on_key(' ', clear_canvas)

print('Click to add dots, press space to clear.')
canvas.run()
```
