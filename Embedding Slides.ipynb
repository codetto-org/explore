# Embedding Slides in Notebooks

**Presentation cells** let you embed a slide deck directly inside a notebook with no separate slideshow app needed. This can be useful to connect your classroom presentation to the notebook activities that students use in an activity.

## Why use a presentation cell?

- **Keeps everything in one place.** No switching between a notebook and a separate slides app.
- **Built with markdown.** If you can write a markdown cell, you already know most of what a slide needs.
- **Click-to-reveal builds.** Bullet points can appear one at a time as you present, keeping a class focused.

## The source format

A presentation cell's source is plain markdown, plus two small extras: optional **frontmatter** at the top, and slides separated by a line containing exactly `---`.

```
---
theme: light
incremental: true
---

# First slide

---

## Second slide
```

Frontmatter fields (all optional):

- `theme`: `light`, `dark`, `sunset`, `ocean`, `forest`, or `slate`. Leave it out and the deck follows whichever theme the app is currently set to.
- `incremental`: `true` or `false` (default `false`). When `true`, each slide's first bulleted or numbered list reveals one item per click instead of appearing all at once.
- `logo`: the filename of an attached image, shown in the bottom-right corner of every slide.
- `background`: a hex color, a CSS gradient, or an attached image's filename.
- `color`: a hex or CSS color, overriding the theme's text color.

## Slide layouts

The very first line of a slide's body can be a directive comment that sets its layout:

```
<!-- layout: title -->
```

- `title`, `section`, and `centered` all center a slide's content — use `title` for a deck's opening slide, `section` for a divider between topics, and `centered` for something like a lone quote or table.
- `columns` splits the slide into two side-by-side zones, at a line containing exactly `:::` (see **Images in slides** below).

When editing a presentation, you can click on the **Insert Slide** button to pick the template that you'd like to use.

## Incremental builds

When `incremental: true` is set (at the deck or slide level), a slide's **first** bulleted or numbered list reveals one item at a time as you click Next, instead of appearing all at once. This can be useful for walking through a list point by point during a lesson.

Clicking **Previous** always jumps straight back a full slide, already fully built — it never un-reveals bullets one at a time. Only the first list on a slide is steppable; content that comes after it appears once the last bullet is revealed.

## Images in slides

You can insert attached image by using its filename, the same as a markdown cell.

Unlike a markdown cell, an image in a slide **fills its slide or column by default** — no extra syntax needed. Add a `{width=NN}` suffix (a percentage, just like markdown cells) for a smaller, proportional size.

## Editing tools

To edit a presentation cell, go into edit mode (the pencil icon on the toolbar), then double-click the deck or click its edit icon. This opens the raw text editor, with the same formatting toolbar as a markdown cell (headings, bold, italic, links, lists, insert image), plus two extra buttons:

- **Insert Slide**: pick a layout from the dropdown and a ready-made slide is inserted at the cursor.
- **Insert Column Break**: inserts the `:::` marker used by a `columns` layout slide.

Close to save; the deck re-parses and shows your changes.

## Navigating and presenting

Students move through the deck with the ◀ ▶ (previous/next) and first/last buttons. Clicking the **Slide N / M** label opens a small popover to jump straight to any slide number.
