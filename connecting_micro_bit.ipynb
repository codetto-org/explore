# Using a micro:bit device within Codetto

A micro:bit is a pocket-sized, programmable computer designed to make learning about coding and electronics simple and fun. You can find out more at https://microbit.org.

Codetto provides full support for deploying code and integrating with micro:bit devices. Beyond deploying code to micro:bit devices, you can also record data from the device and pass other data from the micro:bit to other cells in a notebook.

In this notebook, we'll cover only the basics of connecting to the device.

## Connecting and deploying code

The micro:bit uses a special cell that looks like a regular code cell, but is designed specifically for the micro:bit.

The next cell is a micro:bit cell. Using this cell, you can connect/disconnect from a micro:bit device, deploy code to the micro:bit, and explore the serial monitor - anything that the micro:bit prints back.

The code use a flavor of Python called MicroPython. It looks a lot like the code that you have probably already created in other Codetto notebooks, except it's designed specifically for the micro:bit device.

If you have a micro:bit device, connect it via the USB port now, and click on the **Connect** button in the cell below. Your browser will prompt you to select the device, after which it will be connected.

When you are connected to your micro:bit, then click on the **Deploy** button. This will send the MicroPython code to the device. 

As you can see from the code, it's fairly simple - it runs in a loop and when the either the A or B button is pressed, it plays a sound and writes a message to the console.

After it's deployed to the device, press the A or B button on the micro:bit. Do you hear anything? Then, open the **Serial Monitor** above. This will show you the messages that the micro:bit is sending back.

## Reading micro:bit data back into the notebook

The serial monitor shows you what messages the micro:bit is sending back. It's also possible to bring this data back into the notebook.

The following code cell is a regular Python code cell. It checks to see whether the micro:bit is connected and if it is, it will read all of the messages sent from the micro:bit and display them to the console.

It will continue to listen to messages and print them until you hit the stop button.

```
from codetto import microbit

if not microbit.is_connected():
  print("Connect via the device cell above first")
else:
  while True:
    line = microbit.read_line()
    if line is None:
      print("disconnected")
      break
    print(line)
```

## Where could you take this?

In this notebook, you saw how you can easily connect to a micro:bit from within a Codetto notebook, deploy code to the device, and read it back into the notebook.

While pressing the A or B button is a trivial example, you can hopefully see how easy it would be to collect more realistic data from the micro:bit (e.g., temperature, noise level, etc.), bring it back into the notebook, and display it as a graph.
