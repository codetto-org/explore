# Form Fields

**Form fields** turn the variables at the top of a code cell into text boxes, sliders, dropdowns, and switches. Students change a value using the field and run the code, without having to find and edit the right line. This can be useful for letting students experiment with a program before they're ready to read or change all of its code.

## Why use form fields?

- **Lowers the barrier.** Students can explore what a program does by changing its settings, not its code.
- **Guides experimentation.** A slider's range or a dropdown's choices keep values sensible, so there's less to go wrong.
- **Simple to set up.** A form field is an ordinary Python variable with a short comment on the end of the line.

## Try the example below

Here's a code cell with two form fields, already built. Change the name, move the slider, and click **Run**.

Watch the code as you change a field. The field and the code are two views of the same value, so changing one changes the other.

````python
#@title Say hello
NAME = "Ada" #@param
TIMES = 3 #@param {type:"slider", min:1, max:10, step:1}

for i in range(TIMES):
  print(f"Hello, {NAME}!")
````

## The source format

A form field is a variable assignment, followed by a `#@param` comment on the same line:

```
NAME = "Ada" #@param
```

Because `#@param` is a comment, Python ignores it and the code runs exactly as it would without it. The variable can be used anywhere in the cell, or in any cell that runs afterwards.

A few things to know:

- **The label comes from the variable name.** Underscores become spaces, so `FAVORITE_COLOR` is labelled "Favorite Color". Capital letters are a common way to make these variables stand out, although any name works.
- **The value in the code is the starting value** shown in the field.
- **One variable per line.** The whole assignment and its `#@param` comment must be on a single line.

Adding `#@title` as the very first line of a cell gives the form a heading:

```
#@title Say hello
```

## Field types

What follows `#@param` decides the kind of field:

```
MESSAGE = "Hello" #@param
BORDER = "*" #@param ["*", "=", "~", "+"]
PADDING = 2 #@param {type:"slider", min:0, max:10, step:1}
SHOUT = False #@param {type:"boolean"}
```

- **Text box**: `#@param` with nothing after it. Best for text, since whatever a student types is stored as text.
- **Dropdown**: a list of choices in square brackets. The choices are stored as text too.
- **Slider**: `type:"slider"`, with the smallest value (`min`), the largest (`max`), and the size of each move (`step`). Use a slider whenever you need a number. A `step` such as `0.1` gives decimal values.
- **Switch**: `type:"boolean"`, for a value that's either `True` or `False`.

A dropdown whose choices are color codes, such as `"#ff0000"`, shows a swatch of each color beside it.

Try the example below, which uses all four.

````python
#@title Design a banner
MESSAGE = "Hello, world!" #@param
BORDER = "*" #@param ["*", "=", "~", "+"]
PADDING = 2 #@param {type:"slider", min:0, max:10, step:1}
SHOUT = False #@param {type:"boolean"}

text = MESSAGE.upper() if SHOUT else MESSAGE
space = " " * PADDING
line = BORDER * (len(text) + PADDING * 2 + 2)

print(line)
print(BORDER + space + text + space + BORDER)
print(line)
````

## Hiding the code

Form fields work well with four settings that every code cell has. In edit mode (the pencil icon on the toolbar), they appear as buttons above the cell:

- **Hide code completely**: removes the code editor, leaving only the form and its output.
- **Hide cell controls**: removes the row beneath the form, including the Run button. Only use this with one of the two automatic settings below, as there's otherwise no way to run the cell.
- **Run automatically when a form field changes**: reruns the cell a moment after a field is changed, so there's no need to click Run.
- **Run automatically on load**: runs the cell as soon as the notebook opens, so there's output to see straight away.

Together, these turn a code cell into a small interactive tool. Try the example below, which has all four switched on. Move the slider and the answer updates by itself.

To see the code and controls behind it, go into edit mode and switch off **Hide code completely** and **Hide cell controls**.

````python
#@title Temperature converter
CELSIUS = 20 #@param {type:"slider", min:-40, max:100, step:1}

fahrenheit = CELSIUS * 9 / 5 + 32
print(f"{CELSIUS}°C is {fahrenheit:.1f}°F")
````

## Editing tools

Form fields don't need edit mode or a special cell. Type `#@param` on the end of a variable's line in any code cell and the field appears above the code as you type. Remove the comment and the field goes away.

The four settings described in **Hiding the code** do need edit mode. Switch them on or off with the buttons above a code cell; a button is highlighted while its setting is on.

A cell with a `#@title` is listed under that title in the **Notebook Map** (the panel on the right), which makes it easier to find than "Code cell".

## Coming from Google Colab

Form fields use the same `#@param` and `#@title` comments as forms in Google Colab, so a cell copied from a Colab notebook keeps its fields. The four field types described above are supported. Other Colab field types, such as dates, are shown as a plain text box.
