# Adding Videos to Notebooks

**Video cells** let you place a video directly inside a notebook, right next to the text and code it relates to. This can be useful for introducing a topic, demonstrating a finished project, or walking students through a tricky step before they try it themselves.

## Why use a video cell?

- **Keeps everything in one place.** No sending students to another tab or website to watch a clip.
- **Plays where it's needed.** Put a video immediately before the activity it explains, so students watch and then do.
- **Simple to set up.** A video cell is just a link to a video file, plus a few optional settings.

## The source format

A video cell's source is a small block of settings, written as JSON. Only `url` is required:

```
{
  "url": "/videos/big_buck_bunny_trailer.mp4",
  "title": "Big Buck Bunny trailer",
  "controls": true
}
```

The available settings:

- `url`: the link to the video file. This is the only setting a video cell must have.
- `title`: a short name for the video, shown in the Notebook Map (the panel on the right). Leave it out and the map shows "Video".
- `controls`: `true` or `false` (default `true`). Shows or hides the play, volume, and fullscreen controls. A video that plays automatically always shows them.
- `autoplay`: `true` or `false` (default `false`). Starts the video as soon as the notebook opens.
- `loop`: `true` or `false` (default `false`). Restarts the video each time it reaches the end.
- `captions`: a link to a captions file for the video. See **Adding captions** below.

JSON is strict about punctuation: text values need double quotes, `true` and `false` don't, and every setting except the last ends with a comma.

## Choosing a video

The `url` must be a **direct link to a video file**, the kind of link that usually ends in `.mp4` or `.webm`. Any video file that's publicly available on the web will work, such as one hosted on your school's website.

A link to a YouTube or Vimeo page **won't work**. Those links open a web page with a player on it, rather than the video file itself, and video cells can't embed those players.

For the best results:

- **Use MP4.** MP4 (H.264) videos play in every browser. WebM also works in most.
- **Keep videos short.** A few minutes is plenty, and smaller files start playing sooner on slow connections.
- **Any shape is fine.** The video area is widescreen (16:9). Videos with a different shape are shown in full, with black bars filling the gap.

## Autoplay and looping

Setting `autoplay` to `true` starts a video as soon as the notebook opens. Browsers only allow this for silent videos, so **an autoplaying video is always muted**. Students can unmute it with the controls.

**An autoplaying video always shows its controls**, even if `controls` is set to `false`. A video that starts by itself can be distracting, so students need a way to pause it.

Combine `autoplay` and `loop` and a video plays silently, over and over, like an animation. This suits short clips with no sound, such as a preview of the project students are about to build.

## Adding captions

A video with speech needs **captions** for students who are deaf or hard of hearing. They also help anyone watching in a noisy room. Add them with the `captions` setting:

```
{
  "url": "https://example.com/intro.mp4",
  "title": "Introduction",
  "captions": "https://example.com/intro.vtt"
}
```

The captions are a separate text file in the **WebVTT** format, which usually ends in `.vtt`. Most video editing and captioning tools can save one. The file can be a link, as above, or a file attached to the notebook, such as `"/attachments/intro.vtt"`.

Captions start switched off. Students switch them on from the video's controls, so keep the controls showing on any video that has captions.

To offer captions in more than one language, give a list. Each entry has a link, a language code, and the name students see in the video's captions menu:

```
"captions": [
  { "url": "https://example.com/en.vtt", "lang": "en", "label": "English" },
  { "url": "https://example.com/ja.vtt", "lang": "ja", "label": "日本語" }
]
```

If a captions file can't be loaded, the video still plays without it.

## Editing tools

To add a video cell, go into edit mode (the pencil icon on the toolbar), click **Add cell** where you'd like the video to go, and choose **Video**. The new cell starts with a sample video, ready for you to replace.

To change a video cell, click the edit icon in its top-right corner. This opens the settings as text. Replace the `url` with the link to your own video, adjust any other settings, and click **Close** to save.

If you see **"Enter a valid video URL."** instead of the video, the settings couldn't be read. Check that `url` is present, and look for a missing quote or comma.

## Watching a video

Students play, pause, and skip through a video with the controls along its bottom edge, which also include volume and fullscreen. Each video is listed in the Notebook Map by its `title`, so a clear title makes it easy to find again later.
