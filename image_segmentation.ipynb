# Labelling Every Pixel

If you've used computer vision models before, you'll know that many draw a **box** or a set of **points** around detected objects. The **Image Segmenter** model is unique. Instead, it gives a label to *every single pixel* in the frame. That lets you recolor hair, swap in a background, or change the color of your clothes in real time.

Codetto uses **MediaPipe's Selfie Multi-Class** model, which sorts pixels into six classes:

| Class | What it covers |
|---|---|
| `BACKGROUND` | Everything behind the person |
| `HAIR` | Hair on the head |
| `BODY_SKIN` | Skin on arms and hands |
| `FACE_SKIN` | Skin on the face |
| `CLOTHES` | Clothing |
| `OTHERS` | Glasses, hats, and other extras |

Run the cell below and click **Allow**. Make sure your face, hair, and shoulders are in view. It prints the classes it can see to the console once every 30 frames — press **Stop** when you have seen a few lines.

```python
from codetto import cv, graphics

canvas = graphics.canvas()
camera = cv.start_camera(canvas)
segmenter = cv.start_segmenter(camera)

frame = 0
try:
  while True:
    if frame % 30 == 0:
      segments = segmenter.get_segments()
      print('Classes in view:', segments)
    frame += 1
finally:
  segmenter.stop()
  camera.stop()
```

# Recoloring a Class

`cv.color_segment(canvas, segmenter, class, color, opacity)` fills one class with a solid color. The video still shows through as the color is an overlay laid on top, and `opacity` (from `0.0` to `1.0`) sets how strong it is.

The classes are all derived from the `cv.SEGMENT` constant, for example `cv.SEGMENT.HAIR`.

Run the cell to turn your hair pink. Move your head around and watch the color follow it. Press **Stop** when you are done.

```python
from codetto import cv, graphics

canvas = graphics.canvas()
camera = cv.start_camera(canvas)
segmenter = cv.start_segmenter(camera)

try:
  while True:
    cv.color_segment(canvas, segmenter, cv.SEGMENT.HAIR, '#ff69b4', opacity=0.7)
finally:
  segmenter.stop()
  camera.stop()
```

# Replacing the Background

`cv.apply_image_to_segment(canvas, segmenter, class, image, opacity)` swaps a class for an image, cut exactly to the shape of that class. Point it at `cv.SEGMENT.BACKGROUND` and everything behind you is replaced, while you stay in the picture.

The image here comes from `samples.Images`, the set of pictures built into Codetto. Run the cell to drop yourself in front of a new background. Press **Stop** when you are done.

```python
from codetto import cv, graphics, samples

canvas = graphics.canvas()
camera = cv.start_camera(canvas)
segmenter = cv.start_segmenter(camera)

try:
  while True:
    cv.apply_image_to_segment(canvas, segmenter, cv.SEGMENT.BACKGROUND, samples.Images.Undersea, opacity=0.95)
finally:
  segmenter.stop()
  camera.stop()
```

# Layering Two Effects

You are not limited to one class per frame. Call `cv.color_segment(...)` more than once inside the loop and each call paints its own class. The order matters only where two classes touch.

The cell below colors hair pink and clothes blue at the same time. Press **Stop** when you are done.

```python
from codetto import cv, graphics

canvas = graphics.canvas()
camera = cv.start_camera(canvas)
segmenter = cv.start_segmenter(camera)

try:
  while True:
    cv.color_segment(canvas, segmenter, cv.SEGMENT.HAIR, '#ff69b4', opacity=0.7)
    cv.color_segment(canvas, segmenter, cv.SEGMENT.CLOTHES, '#00bfff', opacity=0.6)
finally:
  segmenter.stop()
  camera.stop()
```

# Your Own Effect

The controls below let you pick the class, the color, and the strength. Change a setting and re-run the cell to see the updated effect.

The class names come through as plain strings, so a small **dictionary** turns the chosen name into the matching `cv.SEGMENT` constant. Press **Stop** when you are done.

```python
from codetto import cv, graphics

CLASS = "HAIR" #@param ["HAIR", "BODY_SKIN", "FACE_SKIN", "CLOTHES"]
color = "#ff69b4" #@param ["#ff69b4", "#00bfff", "#00ff88", "#ff4500", "#ffff00", "#ffffff"]
OPACITY = 0.6 #@param {type:"slider", min:0.1, max:1.0, step:0.1}

segment_for = {
  'HAIR': cv.SEGMENT.HAIR,
  'BODY_SKIN': cv.SEGMENT.BODY_SKIN,
  'FACE_SKIN': cv.SEGMENT.FACE_SKIN,
  'CLOTHES': cv.SEGMENT.CLOTHES,
}

canvas = graphics.canvas()
camera = cv.start_camera(canvas)
segmenter = cv.start_segmenter(camera)

try:
  while True:
    cv.color_segment(canvas, segmenter, segment_for[CLASS], color, opacity=OPACITY)
finally:
  segmenter.stop()
  camera.stop()
```
