# Making Music with Code

A **musical note** is just a sound that vibrates the air at a steady speed. A fast vibration sounds high; a slow one sounds low. That speed is called the **pitch**.

In Codetto you play a note with one line: `audio.play_note(name, seconds)`. Run the cell below to hear a note called `C4` for half a second.

```python
from codetto import audio

audio.play_note('C4', 0.5)
```

# Note Names and Octaves

A note name has two parts: a **letter** from `A` to `G`, then an **octave number**. `C4` means the note C in octave 4, which musicians call *middle C*.

Moving up one octave doubles the pitch, so `C5` sounds twice as fast as `C4` and clearly higher. You can also add a `#` (sharp) or `b` (flat) after the letter, like `F#3` or `Bb5`.

The cell below climbs the same note through three octaves.

```python
for octave in [3, 4, 5]:
  note = f'C{octave}'
  print('playing', note)
  audio.play_note(note, 0.4)
```

# Playing a Chord

A **chord** is several notes sounding at the same moment. To play one, pass a **list** of note names instead of a single name. Every note in the list starts and stops together.

Here is a C major chord: the notes `C4`, `E4`, and `G4` played as one.

```python
audio.play_note('C4', 0.6)          # one note
audio.play_note(['C4', 'E4', 'G4'], 1.0)   # three notes at once
```

# Playing a Melody

A **melody** is notes played one after another. Rather than calling `play_note` over and over, give `audio.play_notes` a list of `(note, seconds)` pairs and it plays the whole tune for you.

An item in the list can be a chord too, by using a list of names in place of a single name.

The cell below plays the opening of *Mary Had a Little Lamb*.

```python
melody = [
  ('E4', 0.4), ('D4', 0.4), ('C4', 0.4), ('D4', 0.4),
  ('E4', 0.4), ('E4', 0.4), ('E4', 0.8),
]

audio.play_notes(melody)
```

# Change the Tune

The list is just data, so your code can build it. The cell below keeps the same seven notes but multiplies every length by a **beat** value you choose with the slider.

Drag the slider and the tune plays again at the new speed. A small beat is fast and lively; a large one is slow.

```python
BEAT = 0.4 #@param {type:"slider", min:0.15, max:0.8, step:0.05}

notes = ['E4', 'D4', 'C4', 'D4', 'E4', 'E4', 'E4']
tune = [(note, BEAT) for note in notes]

audio.play_notes(tune)
print(f'Played 7 notes at {BEAT} seconds each.')
```

# Sound Files and Samples

Notes are not the only sound you can make. `audio.play(path)` plays a **recorded sound file**, like a coin pickup or a laser zap.

Codetto ships a set of ready-to-use files. Instead of typing the file path by hand, import `samples` and let the editor complete the names for you: `samples.Sounds.` lists every short sound effect, and `samples.Music.` lists longer music clips.

Each name is really just the file path as text, so it drops straight into `audio.play`.

```python
from codetto import samples

audio.play(samples.Sounds.CoinPickup)
audio.play(samples.Sounds.Pop)
audio.play(samples.Sounds.Laser)

# The music clips work the same way, e.g. audio.play(samples.Music.Bach)
```

# How the Sound Is Made

Your Python code runs in a background thread that cannot reach the speakers, so `audio.play_note` cannot make sound by itself. Instead it sends your notes across a bridge to the main browser page.

For notes, the page uses the **Web Audio API** built into every browser. It turns each note name into a frequency in hertz with a formula: `A4` is 440 Hz, and each step up the keyboard multiplies the frequency by the twelfth root of two (about 1.0595). It then plays a short **triangle wave** at that frequency, shaped to fade in and out so it sounds a little like a piano key.

For sound files, `audio.play` reads the file from Codetto's built-in file system, sends the bytes to the page, and the page plays them through an ordinary audio player. Either way, nothing is downloaded and no internet is needed.

```python
# A4 is 440 Hz. Each semitone up multiplies by the twelfth root of 2.
step = 2 ** (1 / 12)
print('A4 :', 440)
print('A#4:', round(440 * step, 1))
print('B4 :', round(440 * step ** 2, 1))
print('C5 :', round(440 * step ** 3, 1))
```

# Blocking and Non-Blocking

`audio.play_note`, `audio.play_notes`, and `audio.play` all **block**: the cell stops and waits until the sound has finished before the next line runs. That is what you want when things must happen in order.

Sometimes you want your program to keep going *while* the sound plays, so it can count, animate, or update a score. Every play function has an `_async` twin for that. Two rules:

1. It is an `async` function, so you must put `await` in front of it.
2. It returns straight away, and the rest of your code runs while the sound plays.

```python
await audio.play_notes_async([('C4', 0.3), ('E4', 0.3), ('G4', 0.3), ('C5', 0.6)])

for n in range(1, 8):
  print('counting while the notes play...', n)
```
