# Exploring Data with Pandas

In this notebook you'll learn how to work with structured data using **Pandas**, a Python library for working with structured data, such as spreadsheets or tables. This is one of the most widely used tools in data science and AI.

# Hello, Pandas!

As you start to learn pandas, the key building block is a **DataFrame**. Think of a DataFrame as a table with rows and columns, just like a spreadsheet.

Let's create a DataFrame using average monthly temperatures in New York City:

````python
import pandas as pd

data = {
  'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
  'avg_temp_c': [0, 1, 6, 12, 18, 23, 26, 25, 21, 15, 9, 3]
}

df = pd.DataFrame(data)
df
````

## What is a Series?

A **Series** is a single column of data, like one column pulled from your DataFrame.

You can access a column by name using square brackets `[]`. Let's look at just the temperature column:

````python
temperatures = df['avg_temp_c']
print(temperatures)
````

## Exploring Your Data

Pandas has some handy methods for getting a quick look at your data:

- `df.head()` — shows the first 5 rows
- `df.describe()` — shows summary statistics like min, max, and average
- `df.shape` — tells you how many rows and columns you have

Run each cell below to see what they do:

````python
df.describe()
````

````python
print(f"This dataset has {df.shape[0]} rows and {df.shape[1]} columns.")
````

# Filtering Data Using Pandas

One of Pandas' most useful features is **filtering**, which enables you to select only the rows that match a condition.

The syntax looks like this:

```python
df[df['column_name'] > value]
```

Think of it as asking: *"Give me all the rows where this column meets this condition."*

Let's find all months where the average temperature is above 15°C:

````python
warm_months = df[df['avg_temp_c'] > 15]
warm_months
````

You can use any comparison operator:

- `>` greater than
- `<` less than
- `==` equal to (note the double equals!)
- `>=` greater than or equal to

Your turn! The following code filters the months where the temperature is below 5°C.

Try changing the number oflip the `<` to a `>` to find warmer months instead.

````python
cold_months = df[df['avg_temp_c'] < 5]
cold_months
````

# Sorting Data

You can sort a DataFrame by any column using `df.sort_values()`.

By default it sorts from smallest to largest. Try switching `ascending=False` to flip the order:

````python
df.sort_values('avg_temp_c', ascending=True)
````

# Loading a CSV

So far you've created DataFrames by typing the data directly into Python. In the real world, data usually lives in a file, most commonly a **CSV (comma-separated values)** file.

A CSV file looks like this:

```
month,avg_temp_c,precipitation_mm,sunny_days
Jan,0,94,8
Feb,1,88,9
...
```

Each row is a new line and values are separated by commas. Loading it into Pandas takes just one line:

```python
df = pd.read_csv('/attachments/nyc_weather.csv')
```

(The `attachments` directory specifies files that are attached to this notebook.)

````python
df_weather = pd.read_csv("/attachments/nyc_weather.csv")
df_weather
````

What new columns does the CSV have? Remember that you can use the `describe` method.

````python
df_weather.describe()
````

Your turn! Try filtering or sorting by one of the new columns. This next code filters on precipitation over 100mm.

Can you find months with less than 80mm of precipitation, sorted by the temperatue?

````python
wet_months = df_weather[df_weather['precipitation_mm'] > 100]
wet_months
````
