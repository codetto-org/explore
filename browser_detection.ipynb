# Browser Detection

Not every browser supports every feature. Web Bluetooth (used by the heart rate monitor in `codetto.hardware`) only works in Chrome and Edge. Some webcam and speech features can behave differently across browsers too.

The `codetto.browser` module lets your code check which browser and operating system it's running in. This enables students to have a clear message instead of a confusing error.

## What `codetto.browser` tells you

- `name` - the browser name, e.g. `"Chrome"`, `"Safari"`, `"Firefox"`, `"Edge"`
- `version` - the full version string, e.g. `"128.0.6613.120"`
- `major_version` - just the major version number, e.g. `128`
- `os_name` / `os_version` - the operating system, e.g. `"macOS"` / `"10.15.7"`
- `is_mobile` - `True` if running on a phone or tablet
- `is_chrome`, `is_safari`, `is_firefox`, `is_edge` - convenience `True`/`False` checks

This information comes from the browser itself the moment the notebook loads - there's nothing to configure.

## Try it

Run the cell below to see what `codetto.browser` reports for the browser you're using right now.

```python
from codetto import browser

print("Browser:", browser.name, browser.version)
print("Major version:", browser.major_version)
print("Operating system:", browser.os_name, browser.os_version)
print("Mobile device:", browser.is_mobile)
```

## Writing browser-specific code

Some features only work in certain browsers. For example, Web Bluetooth (used by `codetto.hardware`) only works in Chrome and Edge. Instead of letting students hit a confusing error, check first and show a friendly message:

```python
from codetto import browser

if browser.is_chrome or browser.is_edge:
  print("This browser supports Web Bluetooth - you're good to go!")
else:
  print(f"Heads up: this feature needs Chrome or Edge. You're using {browser.name}.")
```
