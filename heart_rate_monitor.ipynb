# Heart Rate Monitor

In this notebook, you'll explore how Codetto's hardware library can connect to a local BLE (Bluetooth Low Energy) heart rate monitor - e.g., a nearby fitness tracker - and stream heart rate data to a code cell.

```python
#@title First, connect to Bluetooth Heart Rate Monitor
from codetto import hardware

hrm = hardware.HeartRateMonitor.connect() # You'll be prompted to pair with your heart rate monitor here
print('Connected:', hrm.is_connected())
```

```python
#@title Display Live HR Overview
from codetto import graphics
import time

canvas = graphics.canvas()

def displayOverlay(canvas, hr):
  w = canvas.get_width()
  h = canvas.get_height()
  ctx = canvas.get_context('2d')
  ctx.fill_style = "rgba(0, 0, 0, 0.5)"
  ctx.fill_rect(0, 0, w, h)
  ctx.font = "bold 200px sans-serif"
  ctx.fill_style = "white"
  ctx.text_align = "center"
  ctx.text_baseline = "middle"
  ctx.fill_text(str(f"♡ {hr}"), w / 2, h / 2)

try:
  while True:
    reading = hrm.get_reading()
    if reading:
      canvas.clear()
      displayOverlay(canvas, reading.bpm)
    else:
      canvas.clear()
      displayOverlay(canvas, "No reading")
    time.sleep(1)
finally:
  camera.stop()
```
