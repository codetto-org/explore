# Visualizing Data with Matplotlib

Numbers in a table are useful, but charts make patterns *much* easier to see. In this notebook you'll learn how to turn data into visualizations using a popular Python library called **Matplotlib**.

## Setting Up

Before we start using Matplotlib, let's create a Pandas DataFrame of NYC temperature data. We'll use the same data to explore different types of charts. If you are new to Pandas, search for "pandas" in the explore tab to find out more about the library.

```python
import pandas as pd

data = {
  'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
         'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
  'avg_temp_c': [0, 1, 6, 12, 18, 23, 26, 25, 21, 15, 9, 3]
}

df = pd.DataFrame(data)
print("DataFrame loaded")
```

# Line Charts

A **line chart** is ideal for showing how something changes over time, such as temperature across the months of a year.

Let's plot our temperature data. Notice how we import `pyplot` as `plt` from the `matplotlib` package. This is a very common pattern that you'll see with many other chart types.

In this code, we create a figure, plot the temperature, add a title and axis labels, and a background grid to make reading the data easier. Finally we use `plt.show()` to display the chart.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(df['month'], df['avg_temp_c'], marker='o', color='steelblue', linewidth=2)
plt.title('Average Monthly Temperature — New York City')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## Styling Your Chart

Every chart so far uses the same three ingredients: a **color**, a **line style**, and a **marker** for each data point. Mixing and matching these makes a chart easier to read.

| Setting | Some values to try |
|---|---|
| `color` | `'steelblue'`, `'coral'`, `'green'`, `'crimson'`, `'#ff9900'` |
| `linestyle` | `'-'` solid, `'--'` dashed, `':'` dotted, `'-.'` dash-dot |
| `marker` | `'o'` circle, `'s'` square, `'^'` triangle, `'*'` star, `'.'` dot |

One more handy trick: when the x-axis labels are long or crowded, rotate them with `plt.xticks(rotation=45)` so they don't overlap.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(df['month'], df['avg_temp_c'],
     color='crimson', linestyle='--', marker='s', linewidth=2)
plt.title('Same Data, Different Style')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.xticks(rotation=45)
plt.grid(True, linestyle=':', alpha=0.5)
plt.show()
```

## Explore Different Cities

Use the dropdown below to pick a different city and see how the temperature pattern changes!

```python
CITY = "Sydney" #@param ["New York", "Sydney", "London", "Nairobi"]

city_data = {
  "New York":  [0,  1,  6, 12, 18, 23, 26, 25, 21, 15,  9,  3],
  "Sydney":    [26, 26, 24, 21, 18, 15, 14, 15, 17, 20, 22, 25],
  "London":    [6,  6,  9, 12, 15, 18, 20, 20, 17, 13,  9,  6],
  "Nairobi":   [22, 23, 23, 22, 21, 19, 18, 19, 21, 22, 21, 21]
}

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temps = city_data[CITY]

plt.figure(figsize=(10, 5))
plt.plot(months, temps, marker='o', color='coral', linewidth=2)
plt.title(f'Average Monthly Temperature — {CITY}')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## Comparing Two Cities

Every chart so far has shown **one** line. To compare data sets, plot more than one line on the same axes and call `plt.legend()` so the reader knows which is which. The `label=` you give each line is the text that shows up in the legend.

What do you notice about the two lines below?

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
      'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
new_york = [0, 1, 6, 12, 18, 23, 26, 25, 21, 15, 9, 3]
sydney   = [26, 26, 24, 21, 18, 15, 14, 15, 17, 20, 22, 25]

plt.figure(figsize=(10, 5))
plt.plot(months, new_york, marker='o', label='New York')
plt.plot(months, sydney, marker='o', label='Sydney')
plt.title('New York vs. Sydney — Monthly Temperature')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

# Bar Charts

A **bar chart** is great for comparing values side by side. Let's show the same data as bars, and color any cold months (below 10°C) differently.

To do this, we use a special Python construct called a **list comprehension**. 

`colors = ['green' if t >= 10 else 'slateblue' for t in df['avg_temp_c']]`

It's a concise way of look at each value in a series and returning a new list. In this example, the list contains "green" if the temperature is above or equals 10 and "slateblue" otherwise. 

Try changing the temperature values and colors to see different outputs.

```python
colors = ['green' if t >= 10 else 'slateblue' for t in df['avg_temp_c']]

plt.figure(figsize=(10, 5))
plt.bar(df['month'], df['avg_temp_c'], color=colors)
plt.title('Average Monthly Temperature — New York City')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.axhline(y=0, color='black', linewidth=0.8)
plt.show()
```

## Two Charts / Custom Axes

`plt.subplot(rows, cols, position)` places more than one chart in the same figure. Below, the *same* four numbers are drawn twice: the left chart's y-axis starts at 0, the right one starts at 20 (set with `plt.ylim`).

Where the y-axis starts changes how big a difference *looks* — a small gap can be made to look huge. When you read a bar chart, always check the axis.

```python
cities = ['Oslo', 'Berlin', 'Rome', 'Cairo']
july_temp = [22, 25, 30, 35]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.bar(cities, july_temp, color='teal')
plt.title('Y-axis starts at 0')
plt.ylabel('July temp (°C)')

plt.subplot(1, 2, 2)
plt.bar(cities, july_temp, color='teal')
plt.ylim(20, 36)
plt.title('Y-axis starts at 20')

plt.tight_layout()
plt.show()
```

# Histograms

A **histogram** answers a different question: *how are the values spread out?* It sorts the numbers into ranges (called **bins**) and draws a bar showing how many values fall in each range.

Below are the test scores for a class of 40 students. Most scored in the 70s and 80s, with a few at each extreme. Try changing `bins=8` to `bins=4` or `bins=20` and see how the shape changes.

```python
scores = [
  58, 62, 65, 67, 68, 70, 71, 72, 72, 73,
  74, 75, 75, 76, 77, 78, 78, 79, 80, 80,
  81, 82, 82, 83, 84, 85, 85, 86, 87, 88,
  55, 61, 90, 92, 66, 74, 79, 83, 69, 95
]

plt.figure(figsize=(10, 5))
plt.hist(scores, bins=8, color='mediumpurple', edgecolor='white')
plt.title('Distribution of Test Scores')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.show()
```

# Scatter Plots

A **scatter plot** puts one variable on the x-axis and another on the y-axis, with one dot per data point. It's the fastest way to see whether two variables move together.

For example: if students study more do they tend to score higher? If they do, the dots on a scatterplot will form an upward slope. If there's no relationship, the dots will be scattered randomly.

Let's plot hours studied against exam scores for a class of students:

```python
study_data = {
  'hours_studied': [1, 2, 2, 3, 4, 4, 5, 6, 6, 7, 7, 8, 8, 9, 10],
  'exam_score':    [42, 50, 55, 58, 65, 63, 70, 72, 75, 78, 82, 85, 88, 91, 95]
}
df_study = pd.DataFrame(study_data)

plt.figure(figsize=(8, 5))
plt.scatter(df_study['hours_studied'], df_study['exam_score'],
      color='steelblue', s=80, alpha=0.8)
plt.title('Hours Studied vs. Exam Score')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```

## Adding a Line of Best Fit

The dots slope upward, but we can make the trend explicit by drawing the straight line that comes closest to all of them — a **line of best fit**. NumPy's `polyfit` works out that line for us.

```python
import numpy as np

hours = df_study['hours_studied']
exam_scores = df_study['exam_score']

# Fit a straight line (degree 1) through the points
slope, intercept = np.polyfit(hours, exam_scores, 1)
line_y = slope * hours + intercept

plt.figure(figsize=(8, 5))
plt.scatter(hours, exam_scores, color='steelblue', s=80, alpha=0.8)
plt.plot(hours, line_y, color='crimson', linewidth=2, label='Line of best fit')
plt.title('Hours Studied vs. Exam Score')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()

print(f'Each extra hour of study is worth about {slope:.1f} more points.')
```
