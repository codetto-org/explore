# Running Python Code

Every notebook in Codetto is built from **cells**, and the most important kind is the **code cell** - a small block of real, runnable Python. This guide walks through the basics of running Python code in Codetto: running a cell, reading its output, sharing variables between cells, and understanding what happens when code goes wrong.

## Anatomy of a code cell

A code cell has three parts:

- **Editor** - the box where the Python source lives. Click into it to edit.
- **Run button (▶)** - executes just that cell. It turns into a **Stop button (■)** while the cell is running.
- **Output area** - appears below the editor once the cell has run, showing printed text, a result, or an error, in separate tabs.

Try running the cell below.

````python
greeting = "Hello, Codetto!"
print(greeting)
````

### What just happened

Clicking Run sent the code to Python, which printed the greeting to the **Console Tab**. Anything you output with `print()` (or other commands that write to the console) ends up here.

## Console text and output

A cell doesn't only show what you `print()`. If the very last line of a cell is an expression rather than a `print()` call, Codetto shows its value in a separate **Output tab**. Run the cell below to see both at once.

````python
print("This line is printed, so it shows in the Stdout tab.")

3 * 7
````

Notice the two tabs above the output: **Console** for the printed line, and **Output** for the value of `3 * 7`. Only the *last* line of a cell can produce a result this way. Anywhere else in the cell, a bare expression like `3 * 7` does nothing. 

If you've used Jupyter notebooks or Google Colab before, this should feel fairly familiar.

## All cells in a notebook share one Python session

This is the single most important thing to understand about notebooks: every code cell shares the **same** Python session. A variable, function, or import defined in one cell is still available in every other cell and you don't need to repeat yourself.

Run the two cells below, in order.

````python
favorite_subject = "Science"
student_count = 24
````

````python
print(f"This class has {student_count} students studying {favorite_subject}.")
````

The second cell used `favorite_subject` and `student_count` even though neither was defined in it. (They carried over from the cell above.) This is what makes notebooks great for teaching step by step: introduce one idea per cell, and later cells can build on it.

## Order matters: It's *run* order, not page order

Cells share state based on the order you **run** them in, not the order they appear on the page. If you skip a cell, run cells out of sequence, or re-run an earlier cell after changing a later one, the notebook's state can drift away from what a student would expect just by reading top to bottom.

If you have a notebook where you need to run all of the code cells in order (from top to bottom), you can use **Run All Cells**, available in **Cell Actions** menu in the toolbar at the top of the page. It runs every code cell fresh, in page order, so the whole notebook's state always matches what's written on the page.

# When Code Goes Wrong: Errors

Mistakes happen! When a cell hits a problem, Python raises an **exception**, and Codetto shows it in a red **Error tab** instead of Console or Output. Nothing else in the notebook is harmed; only that cell failed.

Run the cell below to see one.

````python
print("Let's calculate a grade average...")

grades = [88, 92, 79]
average = total_grades / len(grades)
print(f"Average: {average}")
````

### Reading a traceback

That block of red text is called a **traceback**. It looks intimidating, but most of the time you only need the **last line** which names the exception type and the specific problem:

```
NameError: name 'total_grades' is not defined
```

Here, `total_grades` was used before it was ever created. This is a classic typo, or a cell run out of order (see the previous section!). The lines above the last one show *where* in the code it happened, useful once you're comfortable reading them, but the last line is always the best place to start.

## Fixing and re-running

Errors aren't permanent. Fix the code and click Run again. The cell picks up right where the notebook left off. Try it yourself: in the cell above, change `total_grades` to `sum(grades)`, then run it again.

## Stopping runaway code

Not every problem shows up as an error. Sometimes code just runs for a long time. This could be an infinite loop, or a slow calculation. While a cell is running, its Run button becomes a **Stop button (■)**. Click it to interrupt execution immediately; whatever had already printed stays in the output, and no error is shown.

Try it below: let it run for a few seconds, then click **Stop** partway through.

````python
import time

for i in range(1, 11):
  print(f"Working... step {i} of 10")
  time.sleep(1)

print("Finished without being stopped!")
````

If you stopped it partway, execution halted right there. No error, no crash, and everything printed up to that point stays visible. This is the tool to reach for any time code seems stuck.

If you are unsure which code cell you should stop, you can also select **Stop All Cells** from the **Cell Actions** menu in the top toolbar.
