# Using Environment Variables

An **environment variable** is a named value that lives *outside* your program's code. Your code can read it while it runs, but the value itself is stored by the environment your program runs in - i.e., it's not stored directly in the notebook.

Think of it like a labelled box that someone else fills in. Your code just asks: *“what's in the box called `API_KEY`?”*

## Why do they matter?

- **Keep secrets out of your code.** API keys, passwords, and tokens should never be pasted into a cell as anyone you share the notebook with would see them. An environment variable keeps the secret on *your* machine only.
- **Change settings without changing code.** The same notebook can behave differently on different computers just by giving it a different value.
- **It's how real software works.** Professional apps read configuration from environment variables all the time. Learning this now is a real-world skill.

# Environment Variables in Codetto

In Python, environment variables are read through the built-in `os` module:

- `os.environ` — a dictionary-like object holding every environment variable.
- `os.getenv("NAME")` — returns the value, or `None` if it isn't set.
- `os.getenv("NAME", "fallback")` — returns `"fallback"` when the variable isn't set, so your code doesn't crash.

Codetto runs Python entirely in your browser, so there's no server operating system underneath. Instead, Codetto gives you your own private list of environment variables that you manage on the **Settings** page. Whatever you add there is loaded into `os.environ` before your code runs.

# Reading an Environment Variable

Even before you add any of your own, the Python environment has a few variables set for you. Run the cell below to look at them.

````python
import os

for name, value in sorted(os.environ.items()):
    print(f"{name} = {value}")
````

Next, we'll use the `os.getenv` to read and print one specific variable from that list. Try changing "HOME" to something different.

````python
home = os.getenv("HOME", "(not set)")
print("HOME is:", home)
````

Finally, let's try to use `os.getenv` to read an environment variable that doesn't exist to see how the fallback works:

````python
missing = os.getenv("THIS_DOES_NOT_EXIST", "nothing here yet")
print("THIS_DOES_NOT_EXIST is:", missing)
````

# Creating your own environment variable

Now let's add one of your own. Follow these steps:

1. Open the **Settings** page (the gear / settings link in the left navigation).
2. Scroll to the **Environment Variables** section.
3. Click **Add Variable**.
4. In the **Name** field, type `MY_NAME` (environment variable names are conventionally written in `UPPER_CASE`).
5. In the **Value** field, type your first name.
6. Click **Add Variable** again to save it. It will now appear in the list, and you can remove it any time with **Delete**.
7. Come back to this notebook.

That value is stored only in this browser, on this device. It is **not** saved inside the notebook file, so it stays private even if you export or share the notebook.

## Reading the variable you just created

Once you've added `MY_NAME` in Settings, run the cell below. Codetto loads your Settings variables into `os.environ` automatically, so no extra setup is needed here.

If it still says the variable isn't set, re-run the cell and double-check the name is exactly `MY_NAME`.

````python
import os

my_name = os.getenv("MY_NAME")

if my_name is None:
    print("MY_NAME is not set yet — add it on the Settings page, then run this cell again.")
else:
    print(f"Hello, {my_name}! Your notebook read that from an environment variable.")
    print(f"Your name has {len(my_name)} letters in it.")
````

# API keys

The most common use is holding a secret key for an online service. Instead of writing:

```python
api_key = "sk-abc123-my-real-secret-key"   # ❌ never do this in a shared notebook
```

you store the key as an environment variable called something like `OPENAI_API_KEY` on the Settings page, and your code reads it:

```python
api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
  raise RuntimeError("Set OPENAI_API_KEY in Settings first")
```

Now the notebook can be shared freely — each person adds their *own* key in their *own* Settings.
