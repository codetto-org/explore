# Welcome to Codetto! 👋

Codetto is a new platform to learn computer science, right in your browser. Codetto is based on notebooks - interactive activities and lessons that combine curriculum, runnable code, videos, quizzes and more. There's nothing to install, and your work saves automatically as you go.

This short guide shows you where everything lives and what to try first.

## Finding your way around

Use the menu on the left to move between the main areas of Codetto:

- **[Home](#/)**: your own notebooks. Anything you open, create, or download shows up here, along with your progress. This is where you'll spend most of your time.
- **[Explore](#/explore)**: a browsable library of ready-made notebooks. Filter by topic, tag, or language and open whatever looks interesting.
- **[Courses](#/courses)**: notebooks grouped into guided, step-by-step paths. A course keeps track of how far you've gotten and points you to the next lesson.
- **[Settings](#/settings)**: change the language and theme, and adjust a few preferences for the code editor.

Once you're inside a notebook, the panel on the right is the **minimap** — it lists every section so you can jump around a long notebook quickly.

## Where to go next

**If you're a student:** head to the **[Explore](#/explore)** tab and open a notebook that catches your eye. Read a bit, run the code, change it, and see what happens — that's the whole idea!

**If you're a teacher:** open the **[Courses](#/courses)** tab and start the **Codetto for Teachers** course. It walks through how notebooks work, how to build your own, and how Codetto can fit into your classroom.
