# Creating QR Codes

QR (Quick Response) codes are those black-and-white square patterns you scan with a phone camera to open a link, join a WiFi network, or save a contact. In this notebook you'll learn what's actually inside a QR code, then generate your own using the **segno** Python library.

## What is a QR code?

A QR code stores data — usually text or a URL — as a grid of black and white squares called **modules**. Three large squares in the corners, the *finder patterns*, let a scanner instantly figure out the code's position and orientation, even photographed at an angle.

Unlike a barcode, which only reads in one direction, a QR code packs data across two dimensions, which is why it can hold far more information in the same amount of space.

## Versions and types

QR codes come in **40 versions**, numbered 1 to 40. Version 1 is a compact 21×21 grid of modules; version 40 is a dense 177×177 grid that can hold thousands of characters. The more data you encode, the higher the version needed to fit it.

There's also a smaller **Micro QR** format (versions M1–M4) for very short payloads. This is useful when space is tight, but comes at the cost of needing a Micro-QR-aware scanner.

You don't usually pick a version yourself. The `segno` Python library automatically chooses the smallest version that fits your data.

## Error correction levels

Every QR code stores redundant data so it can still be scanned even if part of it is smudged, torn, or covered by a logo. There are four levels:

| Level | Recovery capacity |
|---|---|
| `L` (Low) | ~7% |
| `M` (Medium, the default) | ~15% |
| `Q` (Quartile) | ~25% |
| `H` (High) | ~30% |

Higher levels survive more damage, but need more modules to store the same data and the code's image needs to be denser. `H` is a common choice when a logo will sit in the middle of the code.

# Creating your first QR code

segno's `make()` function takes any text and returns a QR code object. Run the cell below to make your first QR code!

```python
import segno

qr = segno.make("https://codetto.app")
qr
```

Try scanning the QR code above with your phone's camera. It should open codetto.app on your phone's browser. (PS: Codetto notebooks also run on smartphones!)

# Customizing your QR code

The `segno` library gives you control over how much damage it can survive and how it looks:

- `error` — error correction level (`"L"`, `"M"`, `"Q"`, `"H"`)
- `scale` — how many pixels wide each module is (bigger = larger image)
- `border` — the quiet zone (blank margin) around the code, in modules
- `dark` / `light` — colors for the modules and background

```python
qr = segno.make("Codetto", error="H")
qr.show(scale=10, dark="#8e24aa", light="#f3e5f5", border=2)
```

### Try it yourself

Change the fields below and re-run the cell. The code is set to run automatically whenever a field changes, so you'll see the QR code update live as you experiment.

# Using QR codes for contact cards and WiFi networks

Beyond plain text and links, QR codes can encode structured data using standard formats. For example, a phone camera recognizes the format and offers to *join this network* or *save this contact* instead of just opening a webpage. The segno library builds these payloads for you via `segno.helpers`.

### WiFi network

`helpers.make_wifi()` builds a QR code that joins a WiFi network when scanned — handy for a classroom or guest network sign.

```python
from segno import helpers

wifi_qr = helpers.make_wifi(ssid="Classroom-WiFi", password="letmein123", security="WPA")
wifi_qr
```

`security` accepts `"WPA"` (covers WPA/WPA2/WPA3-personal), `"WEP"`, or `None` for an open network with no password.

### Contact card (vCard)

`helpers.make_vcard()` builds a QR code that saves a contact straight to a phone's address book.

```python
contact_qr = helpers.make_vcard(
    name="Guest;Codetto",
    displayname="Codetto Guest",
    email="hello@codetto.app",
    url="https://codetto.app",
    org="Codetto",
)
contact_qr
```
