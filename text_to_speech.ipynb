# Talking Computers

**Text to speech** (TTS) turns written words into spoken audio. You have heard it many times: a screen reader reading a web page aloud, a voice assistant like Siri or Alexa, the turn-by-turn voice in a maps app, an audiobook, or the announcements at a train station.

In Codetto you make the computer talk with one line: `audio.speak(...)`. Run the cell below to hear it.

```python
from codetto import audio

audio.speak("Hello! I am your computer, and I can talk.")
```

# From Robots to Neural Voices

Early text-to-speech, from the 1980s through the 2000s, was **rule-based** or stitched together from tiny recorded fragments of a human voice. It worked, but it sounded flat and robotic.

Today's TTS uses **neural networks** trained on many hours of recorded speech. The computer learns the rhythm, stress, and melody of real talking, so a modern voice sounds much closer to a person. Listen to the difference in a full sentence:

```python
audio.speak(
  "Modern text to speech can pause at commas, raise its pitch for a question, "
  "and stress the important words, just like a person reading aloud."
)
```

# How Codetto Speaks

Your Python code runs in a background thread that has no access to the screen or the speakers. So `audio.speak` cannot make sound by itself.

Instead it sends your text across a bridge to the main browser page, which calls the **Web Speech API** built into every modern browser (`speechSynthesis`). That API then asks the **operating system** to synthesise the audio and play it.

A few things follow from this:

- No sound files and no internet are involved. The audio is generated on your device as it plays.
- The available voices are the ones installed on *your* computer, so Codetto does not ship them.
- If a browser has no speech support at all, `audio.speak` prints a message and does nothing, rather than crashing.

```python
audio.speak("My voice is made by your device, using the speech engine built into the browser.")
```

# Choosing a Voice

Pass a `voice` to pick a language and accent. Voices are named `audio.Voice.<LOCALE>.<GENDER>`, where `<GENDER>` is `FEMALE` or `MALE`.

Available locales:

`EN_US` `EN_GB` `EN_AU` `FR_FR` `DE_DE` `ES_ES` `IT_IT` `PT_BR` `JA_JP` `ZH_CN` `KO_KR` `HI_IN` `AR_SA`

Choosing a voice also sets the language, so the words are pronounced with that accent.

```python
line = "Hello! This is how I sound."

audio.speak(line, voice=audio.Voice.EN_US.FEMALE)
audio.speak(line, voice=audio.Voice.EN_GB.MALE)
audio.speak(line, voice=audio.Voice.EN_AU.FEMALE)
```

# Every Device Is Different

Because the voices come from your operating system, the exact set is different on a Mac, a Windows PC, and a Chromebook.

For each language and gender, Codetto keeps a list of good voice names and uses the first one that is installed. If none of them are found, it falls back to *any* voice for that language. That means the same code can sound a little different from one computer to the next, and once in a while a requested voice is simply not available.

Pick a voice below and run the cell to hear it. Change the dropdowns and it speaks again.

```python
LOCALE = "en-US" #@param ["en-US", "en-GB", "en-AU", "fr-FR", "de-DE", "es-ES", "it-IT", "pt-BR", "ja-JP", "zh-CN", "ko-KR", "hi-IN", "ar-SA"]
GENDER = "female" #@param ["female", "male"]

locale = getattr(audio.Voice, LOCALE.replace("-", "_").upper())
chosen = locale.FEMALE if GENDER == "female" else locale.MALE

audio.speak("Hello from your chosen voice.", voice=chosen)
print(f"Spoke with {LOCALE} ({GENDER}).")
```

# Waiting for the Words: audio.speak

`audio.speak(text)` **blocks**. The cell stops and waits until the whole sentence has been spoken, and only then runs the next line.

That is what you want when things must happen in order. Watch how the messages below appear in step with the speech.

```python
audio.speak("First, I say this.")
print("Now the first sentence is finished.")

audio.speak("Then, I say this.")
print("And now the second one is finished.")
```

# Carrying On: audio.speak_async

Sometimes you want the program to keep going *while* the computer talks: animate something, update a score, count down. Use `audio.speak_async(text)`.

Two rules:

1. It is an `async` function, so you must put `await` in front of it.
2. It returns straight away. The speech starts, and the rest of your code runs at the same time.

If you call it several times, the sentences line up and play one after another.

```python
await audio.speak_async("I will keep talking while the numbers count up.")

for n in range(1, 8):
  print("working...", n)
```
