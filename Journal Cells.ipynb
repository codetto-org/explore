# Journal Cells

**Journal cells** give students a place to write inside a notebook. You provide a prompt, and each student writes their own response beneath it. This can be useful for reflections, predictions, and explanations, where you'd like students to put their thinking into words.

## Why use a journal cell?

- **Keeps everything in one place.** Students write their notes right next to the activity they're about, not in a separate document.
- **Open-ended.** There's no right or wrong answer, so students can explain, predict, or reflect in their own words.
- **No edit mode needed.** Students write straight into the journal, and their work is saved with the notebook.

## The source format

A journal cell's source has two parts, separated by a line containing exactly `---`:

```
What surprised you about today's activity?

---

The student's response goes here.
```

- **Above the line** is the **prompt**. You write this, and it stays fixed while students write.
- **Below the line** is the **response**. This starts out empty and is filled in by the student.

When writing a prompt, leave everything below the `---` empty. Only the first `---` line counts as the separator, so don't use one inside the prompt itself.

## Formatting with markdown

Both parts of a journal are markdown, the same as a markdown cell. A prompt can use **bold**, *italic*, lists, links, and more, which is handy for breaking a larger task into steps.

Students can use markdown in their responses too, although plain sentences work perfectly well.

## Journals without a prompt

The prompt is optional. Leave it out and the journal becomes an open space for notes, with no question above it. This suits a "notes" area at the end of a lesson, or anywhere students might want to jot something down.

## Editing tools

To add a journal cell, go into edit mode (the pencil icon on the toolbar), click **Add cell** where you'd like the journal to go, and choose **Journal**. The new cell starts with an empty prompt and an empty response.

To write the prompt, stay in edit mode and double-click the journal, or click its edit icon. This opens the whole cell as text, with the `---` line already in place. Type your prompt above the line, then click **Close** to save.

Outside edit mode, double-clicking a journal only ever opens the response, so students can't change your prompt by accident.

## Saving and sharing responses

Responses are saved with the notebook, so they're still there the next time a student opens it.

When a student exports their notebook, they can choose **Export with my work** to keep their journal responses, for example to hand in. If you've been trying out the journals yourself, choose **Export a clean copy** instead, which clears every response and keeps your prompts. Sharing a notebook as a link always sends a clean copy.
