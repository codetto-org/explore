# Working with Numbers in NumPy

When you have a lot of numbers to crunch, plain Python lists get slow and clumsy. **NumPy** (short for "Numerical Python") gives you a faster, simpler way to store and calculate with numbers. It is the foundation that Pandas, Matplotlib, and almost every data science tool is built on.

In this notebook you'll meet the NumPy **array** and see how it lets you do math on a whole list of numbers at once.

# From Lists to Arrays

The core of NumPy is the **array**: an ordered collection of values, all of the same type. It looks a lot like a Python list, but it can do maths on every value at once.

Let's record how many steps someone walked on each day of one week, and turn that list into an array with `np.array()`.

```python
import numpy as np

steps = np.array([9200, 10500, 8000, 12300, 15000, 6400, 11800])
print(steps)
print(type(steps))
print("Number of values:", len(steps))
print("Type of each value:", steps.dtype)
```

# Math Without Loops

Here is the big idea. To change every value in a plain Python list you need a loop. With a NumPy array you just write the calculation once, and it applies to **every value automatically**. This is called **vectorization**.

An average walking step is about 0.00076 km long, so we can turn step counts into distances in a single line.

```python
# Distance walked each day, in kilometres
distance_km = steps * 0.00076
print(distance_km)

# Arithmetic between two arrays happens value by value too
goal = np.array([10000, 10000, 10000, 10000, 10000, 8000, 8000])
difference = steps - goal
print("Steps above or below goal:", difference)
```

# Summarizing an Array

NumPy arrays come with built-in methods that boil a whole array down to one number. These are called **aggregations**.

- `.sum()` adds every value
- `.mean()` gives the average
- `.min()` and `.max()` find the smallest and largest
- `.std()` measures how spread out the values are

```python
print("Total steps this week:", steps.sum())
print("Average per day:", steps.mean())
print("Best day:", steps.max())
print("Quietest day:", steps.min())
print("Spread (standard deviation):", round(steps.std(), 1))
```

# Picking Out Values

You can read one value, or a range of values, using square brackets. Counting starts at **0**, and negative numbers count back from the end.

The most powerful trick is a **boolean mask**: put a comparison inside the brackets and NumPy keeps only the values that pass the test.

```python
print("First day:", steps[0])
print("Last day:", steps[-1])
print("First three days:", steps[:3])

# Boolean mask: only the days that hit 10,000 steps
big_days = steps[steps >= 10000]
print("Days that reached the goal:", big_days)
print("How many:", (steps >= 10000).sum())
```

# Building Arrays Quickly

Typing out every value gets old fast. NumPy has helpers that build arrays for you:

- `np.arange(start, stop, step)` counts like Python's `range`
- `np.linspace(start, stop, count)` gives evenly spaced numbers, including both ends
- `np.zeros(n)` and `np.ones(n)` fill an array with 0s or 1s

```python
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
print(np.zeros(4))
print(np.ones(3))
```

# 2D Arrays

An array can have rows and columns, like a spreadsheet. Here are the step counts for **two** people across the same week.

`.shape` tells you the size as `(rows, columns)`. When you aggregate, `axis=1` works along each row, and `axis=0` works down each column.

```python
week = np.array([
  [9200, 10500, 8000, 12300, 15000, 6400, 11800],
  [7000,  8200, 9100, 10000,  9900, 12000,  8500]
])

print("Shape:", week.shape)
print("Value at row 0, column 4:", week[0, 4])
print("Total per person:", week.sum(axis=1))
print("Average per day (both people):", week.mean(axis=0))
```

## Random Numbers

`np.random.randint(low, high, size=n)` rolls `n` whole numbers. `randint(1, 7)` gives values from 1 to 6, just like a dice.

Drag the slider to roll more dice. A fair dice averages **3.5**, and the more you roll, the closer the average gets.

```python
NUM_ROLLS = 440 #@param {type:"slider", min:10, max:5000, step:10}

rolls = np.random.randint(1, 7, size=NUM_ROLLS)
print("Rolled", NUM_ROLLS, "dice")
print("Average roll:", round(rolls.mean(), 2))
```

# NumPy Meets Matplotlib

`np.linspace` is perfect for drawing smooth curves: make a lot of evenly spaced x-values, calculate a y-value for each one, and hand both arrays to Matplotlib.

Here we take 100 points from -5 to 5 and square each one to draw a parabola.

```python
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)
y = x ** 2

plt.figure(figsize=(8, 5))
plt.plot(x, y, color='steelblue', linewidth=2)
plt.title('y = x squared, from 100 points')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
